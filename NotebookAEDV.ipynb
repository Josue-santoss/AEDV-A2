## Introdução:

O estado do Rio de Janeiro vivencia, há décadas, um dos cenários de segurança pública mais complexos e militarizados da América Latina. O presente artigo investiga a arquitetura espaço-temporal da violência letal fluminense a partir da integração de 15 bases abertas do Instituto de Segurança Pública (ISP-RJ) e de três malhas geográficas oficiais (CISP, AISP e RISP), cobrindo séries longitudinais entre 1991 e 2026. Estruturado em torno de sete perguntas de domínio em progressão lógica — da descrição macro à análise espaço-temporal e de defasagens —, o estudo examina desde a transformação tecnológica nos arsenais apreendidos até a concentração territorial da letalidade policial, do feminicídio e dos ciclos sazonais.

### Síntese dos Principais Achados
- **Militarização do arsenal (P1):** Entre 2007 e 2024, o total anual de armas de fogo apreendidas caiu **44,4%** (de 11.062 para 6.150; taxa de 71,47 para 35,02/100 mil hab.), enquanto as apreensões de **fuzis mais que triplicaram** (+242,1%, de 214 para 732; fatia saltando de 1,93% para 11,90%), com substituição estrutural de revólveres por pistolas.
- **Geografia distinta entre letalidade geral e feminicídio (P2):** Normalizadas pela população (2017–2022), as maiores taxas de letalidade violenta concentram-se na Baixada Fluminense e polos litorâneos (lideradas por AISP 24 — Itaguaí/Seropédica, com 58,13/100 mil), enquanto **6 das 10 maiores taxas de feminicídio estão no Interior** (correlação espacial baixa, $r = 0{,}354$, e divergência temporal acentuada em 2017–2025, $r = -0{,}869$).
- **Sazonalidade sincronizada (P3):** Ajustada pela duração dos meses (2003–2024), a letalidade violenta apresenta picos consistentes em **fevereiro–março** (índice 112,0 no estado) e vales no inverno (**julho–agosto**, índice 89,8), com maior amplitude no Interior (25,1 pontos).
- **Queda metropolitana e convergência regional (P4):** De 2003 a 2022, a taxa caiu **66,7%** na Capital (58,3 → 19,4/100 mil) e **49,0%** na Baixada Fluminense (70,4 → 35,9/100 mil), frente a **−24,6%** no Interior (34,5 → 26,0/100 mil), reduzindo a razão Máx/Mín regional de 2,04× para 1,85×.
- **Clusters de confronto armado (P5):** Há forte associação espacial entre volume de fuzis apreendidos e taxa de mortes por intervenção policial ($r = 0{,}757$; Spearman $\rho = 0{,}834$), com **15 das 39 AISPs** formando um corredor crítico contínuo na Região Metropolitana (mais Angra dos Reis).
- **Peso da intervenção policial nos grandes municípios (P6):** Entre 2014 e 2024, nos 6 municípios mais populosos, a participação da MDIP chegou a **55,7%** em Niterói (2023) e **48,7%** em São Gonçalo (2021), permanecendo residual em Campos dos Goytacazes (máx. 3,6%); enquanto variações anuais caminham juntas ($r_{\Delta}$ entre $+0{,}22$ e $+0{,}61$), no longo prazo a fatia de MDIP cresceu mesmo com queda da letalidade geral em 4 dos 6 municípios.
- **Vitimização policial e letalidade subsequente (P7):** Ao longo de 284 meses (2003–2026; 596 policiais mortos em serviço), a correlação positiva em nível ($r \approx 0{,}39$ a $0{,}45$) reflete tendência histórica compartilhada; ao remover a tendência em primeiras diferenças, o efeito no mês seguinte desaparece ($r_{\text{lag 1}} = +0{,}005$ para letalidade total e $+0{,}015$ para MDIP), afastando causalidade mecânica no agregado mensal estadual.

   - **AISP:** A Área Integrada de Segurança Pública é uma divisão territorial que alinha a atuação da Polícia Militar (PMERJ) e da Polícia Civil (PCERJ) na mesma região para combater a criminalidade de forma conjunta.

```python
def taxa_por_100_mil(
    quantidade: pd.Series,
    populacao: pd.Series,
    fator: int = FATOR_TAXA,
) -> pd.Series:
    """
    Calcula a taxa de ocorrências por 100 mil habitantes, dividindo
    a quantidade absoluta pela população correspondente e multiplicando
    pelo fator de padronização. Retorna NaN quando a população é
    ausente, nula ou negativa.
    """
    denominador_valido = populacao.where(populacao > 0)
    return quantidade.div(denominador_valido).mul(fator)


def agregar_contagens_aisp_ano(
    base: pd.DataFrame,
    colunas_contagem: list[str],
) -> pd.DataFrame:
    """
    Agrega as colunas de contagem especificadas por ano e AISP,
    somando os registros pertencentes a cada grupo. Verifica a
    existência das colunas obrigatórias e retorna um DataFrame
    com uma única linha por combinação de ano e AISP.
    """
    exigir_colunas(base, ["ano", "id_aisp", *colunas_contagem], "base de ocorrências")
    return (
        base.groupby(["ano", "id_aisp"], as_index=False)[colunas_contagem]
        .sum()
    )
```

## 2. Carregamento e catálogo das bases

Os arquivos CSV são identificados automaticamente a partir da pasta de dados. Novas bases adicionadas com a extensão `.csv` passam a integrar `dataframes` sem alterações no carregador.

```python
def carregar_csvs_do_repositorio(url_zip: str, pasta_dados: str, prefixo: str) -> dict[str, pd.DataFrame]:
    """Lê os CSVs do ZIP do GitHub; mantém cada tabela separada e identificada pelo nome."""
    resposta_zip = requests.get(url_zip, timeout=60)
    resposta_zip.raise_for_status()

    bases_carregadas = {}
    with ZipFile(BytesIO(resposta_zip.content)) as arquivo_zip:
        for caminho in arquivo_zip.namelist():
            if pasta_dados not in caminho or not caminho.lower().endswith(".csv"):
                continue

            nome_arquivo = caminho.rsplit("/", 1)[-1]
            nome_base = nome_arquivo.removesuffix(".csv").removeprefix(prefixo)
            if nome_base in bases_carregadas:
                raise ValueError(f"Nome de base duplicado no ZIP: {nome_base}")

            with arquivo_zip.open(caminho) as conteudo_csv:
                bases_carregadas[nome_base] = pd.read_csv(
                    conteudo_csv, sep=None, engine="python"
                )

    if not bases_carregadas:
        raise ValueError(f"Nenhum arquivo CSV encontrado em {pasta_dados!r}.")
    return bases_carregadas


dataframes_brutos = carregar_csvs_do_repositorio(URL_ZIP_DADOS, PASTA_DADOS_ZIP, PREFIXO_CSV)
print(f"{len(dataframes_brutos)} bases importadas:", ", ".join(sorted(dataframes_brutos)))
```

## 3. Limpeza e padronização

Esta rotina produz **cópias** das bases de entrada. `dataframes_brutos` preserva os CSVs importados; `dataframes` contém as bases tratadas. Corrige textos, preserva valores ausentes de chaves/população e define uma coluna temporal `data` nas tabelas mensais.

**Critério mantido do código original:** ausências em colunas numéricas de contagem iniciadas por `quantidade_` são preenchidas com zero. Essa hipótese deve ser confrontada com o dicionário oficial dos dados; não deve ser automaticamente aplicada às novas métricas.

```python
CORRECOES_REGIAO = {
    "Grande NiterÃƒÂƒÃ‚ÂƒÃƒÂ‚Ã‚Â³i": "Grande Niterói",
    "Grande Niteroi": "Grande Niterói",
}


def exigir_colunas(base: pd.DataFrame, colunas: list[str], nome_base: str) -> None:
    """Interrompe o processamento caso falte alguma coluna esperada."""
    faltantes = set(colunas).difference(base.columns)
    if faltantes:
        raise KeyError(f"{nome_base}: colunas ausentes: {sorted(faltantes)}")


def limpar_bases_isp(bases: dict[str, pd.DataFrame]) -> dict[str, pd.DataFrame]:
    """Padroniza as bases sem alterar os DataFrames de entrada."""
    bases_tratadas = {}

    for nome_base, base_original in bases.items():
        base = base_original.copy()
        base.columns = base.columns.str.strip()

        # Limpeza de textos sem transformar ausências (NaN) em strings 'nan'.
        for coluna in base.select_dtypes(include=["object", "string"]).columns:
            base[coluna] = base[coluna].map(
                lambda valor: valor.strip() if isinstance(valor, str) else valor
            )

        if "regiao" in base.columns:
            base["regiao"] = base["regiao"].replace(CORRECOES_REGIAO)

        # Hipótese de imputação do projeto: somente contagens de ocorrências.
        for coluna in base.select_dtypes(include="number").columns:
            if coluna.startswith("quantidade_"):
                base[coluna] = base[coluna].fillna(0)

        if "ano" in base.columns:
            base["ano"] = pd.to_numeric(base["ano"], errors="raise").astype("int64")
        if "mes" in base.columns:
            base["mes"] = pd.to_numeric(base["mes"], errors="raise").astype("int64")
        if {"ano", "mes"}.issubset(base.columns):
            base["data"] = pd.to_datetime(
                {"year": base["ano"], "month": base["mes"], "day": 1},
                errors="raise"
            )

        bases_tratadas[nome_base] = base

    return bases_tratadas


dataframes = limpar_bases_isp(dataframes_brutos)
print("Bases padronizadas:", len(dataframes))
```

## 5.2. Malhas geográficas (CISP, AISP, RISP)

Centralizamos aqui o carregamento dos shapefiles oficiais (`CISP.zip`, `AISP.zip` e `RISP.zip`) já reprojetados para `EPSG:4326` e com chaves padronizadas (`id_cisp`, `id_aisp`, `id_risp`).

```python
def carregar_shapefile_github(nome_zip: str, chave_padrao: str, tolerancia: float = 0.005) -> gpd.GeoDataFrame:
    """Baixa shapefile compactado do repositório, padroniza CRS (EPSG:4326) e chave territorial."""
    url = f"{URL_BASE_SHAPEFILES}/{nome_zip}"
    resposta = requests.get(url, timeout=60)
    resposta.raise_for_status()

    with tempfile.NamedTemporaryFile(delete=False, suffix=".zip") as tmp:
        tmp.write(resposta.content)
        tmp_path = tmp.name

    try:
        gdf = gpd.read_file(f"zip://{tmp_path}")
    finally:
        os.unlink(tmp_path)

    sufixo = chave_padrao.replace("id_", "")
    candidatos = [chave_padrao, chave_padrao.upper(), sufixo, sufixo.upper()]
    col_encontrada = next((c for c in candidatos if c in gdf.columns), None)
    if col_encontrada is None:
        raise KeyError(
            f"{nome_zip}: nenhuma coluna de chave encontrada entre {candidatos}. Colunas: {list(gdf.columns)}"
        )

    if col_encontrada == chave_padrao:
        gdf[f"{chave_padrao}_original"] = gdf[col_encontrada]
    gdf[chave_padrao] = pd.to_numeric(gdf[col_encontrada], errors="raise").astype("int64")

    if gdf.crs is None:
        raise ValueError(f"{nome_zip}: GeoDataFrame sem CRS definido.")
    if gdf.crs.to_epsg() != 4326:
        gdf = gdf.to_crs(epsg=4326)

    if tolerancia is not None:
        gdf["geometry"] = gdf["geometry"].simplify(tolerance=tolerancia, preserve_topology=True)

    return gdf


gdf_cisp = carregar_shapefile_github("CISP.zip", "id_cisp")
gdf_aisp = carregar_shapefile_github("AISP.zip", "id_aisp")
gdf_risp = carregar_shapefile_github("RISP.zip", "id_risp")

print(f"Shapefiles carregados: CISP={len(gdf_cisp)}, AISP={len(gdf_aisp)}, RISP={len(gdf_risp)}")
```

## 6.1. Dimensão de relacionamento CISP → AISP

A tabela cadastral serve para consultas. Para os indicadores históricos, extraímos da base mensal a AISP registrada para cada combinação de ano, mês e CISP, incluindo meses com zero vítimas. Uma CISP pode mudar de AISP entre meses; duas atribuições no mesmo mês exigem revisão.

```python
df_mapa_cisp_aisp = (
    df_rel[["id_cisp", "id_aisp"]]
    .dropna(subset=["id_cisp", "id_aisp"])
    .drop_duplicates()
    .copy()
)

conflitos_cisp_aisp = (
    df_mapa_cisp_aisp
    .groupby("id_cisp")["id_aisp"]
    .nunique()
    .loc[lambda serie: serie > 1]
)
if not conflitos_cisp_aisp.empty:
    raise ValueError(
        "Uma CISP foi associada a mais de uma AISP. "
        "Não é seguro escolher .first(); verifique a correspondência temporal. "
        f"Exemplos: {conflitos_cisp_aisp.head().to_dict()}"
    )

# Mantido como recurso de consulta, sem ser a fonte primária da relação.
aisp_de_cisp = df_mapa_cisp_aisp.set_index("id_cisp")["id_aisp"].to_dict()

chave_mensal = ["ano", "mes", "id_cisp"]
mapa_cisp_aisp_mensal = df_cisp_mensal[chave_mensal + ["id_aisp"]].drop_duplicates()
if mapa_cisp_aisp_mensal[chave_mensal + ["id_aisp"]].isna().any().any():
    raise ValueError("Há chaves territoriais mensais ausentes.")
if mapa_cisp_aisp_mensal.duplicated(chave_mensal).any():
    raise ValueError("Há uma CISP com duas AISPs no mesmo mês.")

territorio_aisp1 = (
    df_cisp_mensal.loc[df_cisp_mensal["id_aisp"].eq(1)]
    .groupby("ano")
    .agg(cisps=("id_cisp", lambda s: tuple(sorted(s.unique()))),
         meses_observados=("mes", "nunique"))
)
nomes_aisp1 = {
    (6, 7): "Cidade Nova / Santa Teresa (Rio de Janeiro)",
    (72, 73): "São Gonçalo / Neves",
}
territorio_aisp1["nome_aisp"] = territorio_aisp1["cisps"].map(nomes_aisp1).fillna(
    "AISP 1 — composição territorial a verificar"
)
```

Em cada mês, somamos a população das CISPs vinculadas à AISP na base de ocorrências. Um mês só contribui para a média se todas as suas CISPs tiverem população. A estimativa anual é a média desses totais mensais completos. Em 2022, por exemplo, a fonte populacional termina em julho. Registramos separadamente os meses com vínculo territorial e os meses com população completa.

Para comparar taxas anuais, exigimos 12 meses de atividade territorial e população média positiva. Uma AISP ativa apenas durante parte do ano conserva sua contagem e sua estimativa de população, com a taxa anual ausente.

## Estatísticas descritivas reutilizáveis

Este bloco sintetiza a **distribuição entre AISPs**, e não a taxa do estado. A média simples de taxas por AISP dá o mesmo peso a cada AISP; a taxa estadual exigiria somar os eventos e as populações antes da divisão.

## 14. Resumo anual das taxas por AISP

O cálculo do desvio-padrão mantém `ddof=0`. Para ampliar o projeto, a função pode receber outra coluna de taxa e retornar o mesmo conjunto de estatísticas por ano.

```python
def resumir_taxas_por_ano(
    painel: pd.DataFrame,
    coluna_taxa: str = "taxa_letalidade",
) -> pd.DataFrame:
    """Resume a distribuição anual de uma taxa entre AISPs."""
    exigir_colunas(painel, ["ano", coluna_taxa], "painel de taxas")
    resumo = (
        painel.groupby("ano")[coluna_taxa]
        .agg(
            media="mean",
            mediana="median",
            minimo="min",
            maximo="max",
            q1=lambda serie: serie.quantile(0.25),
            q3=lambda serie: serie.quantile(0.75),
            desvio_padrao=lambda serie: serie.std(ddof=0),
        )
        .reset_index()
    )
    resumo["variancia"] = resumo["desvio_padrao"] ** 2
    resumo["amplitude_interquartil"] = resumo["q3"] - resumo["q1"]
    resumo["coef_variacao"] = (
        resumo["desvio_padrao"]
        .div(resumo["media"].where(resumo["media"] != 0))
        .mul(100)
    )
    return resumo


resumo_estatistico = resumir_taxas_por_ano(estatisticas_aisp)
resumo_estatistico.head()
```

# Perguntas:

## Pergunta 1 — Evolução Histórica das Apreensões de Armas de Fogo e Fuzis no RJ

> **Pergunta de domínio:** *Qual é a evolução histórica do volume absoluto e da taxa (por 100 mil habitantes) de armas de fogo e fuzis apreendidos no estado do Rio de Janeiro nos últimos anos?*

### 1.1 Transformação e Preparação

A partir da base mensal já padronizada `df_armas` (`armas_apreendidas_mensal`), filtramos os anos completos da série histórica (**2007 a 2024**) e agregamos anualmente o total de armas de fogo apreendidas (`quantidade_arma_fogo_total`) e os três principais tipos de armas (`quantidade_fuzil`, `quantidade_pistola` e `quantidade_revolver`).

Para separar o efeito do crescimento demográfico da dinâmica criminal, cruzamos as contagens anuais com `populacao_uf_anual` e calculamos as taxas por 100 mil habitantes utilizando a função utilitária `taxa_por_100_mil`.

Durante essa etapa, constatamos a ausência de dados populacionais referentes aos anos de **2023 e 2024** na base utilizada. Para viabilizar o cálculo das taxas nesses anos, aplicamos o método de preenchimento progressivo (*forward fill*, `ffill`), adotando a população de 2022, último ano com informação disponível, como referência para ambos os anos. Dessa forma, as taxas de 2023 e 2024 foram calculadas sob a hipótese de estabilidade populacional em relação a 2022, constituindo uma limitação metodológica da análise.

Também derivamos:
- **Índice Base 100 (2007 = 100)** das taxas por 100 mil habitantes, permitindo comparar diretamente a trajetória relativa do total de armas e dos fuzis em uma mesma escala, sem distorções de eixos duplos;
- **Participação percentual (%)** de revólveres, pistolas e fuzis no total apreendido, para mensurar a mudança qualitativa (substituição tecnológica) do arsenal.

```python
populacao_uf_anual.head()
```

```python
# 1.1 Transformação e preparação — Pergunta 1
colunas_armas_p1 = [
    "quantidade_arma_fogo_total",
    "quantidade_fuzil",
    "quantidade_pistola",
    "quantidade_revolver",
]
exigir_colunas(df_armas, ["ano", *colunas_armas_p1], "df_armas")

armas_anual = (
    df_armas.loc[df_armas["ano"].between(2007, 2024)]
    .groupby("ano", as_index=False)[colunas_armas_p1]
    .sum()
    .sort_values("ano")
    .merge(populacao_uf_anual, on="ano", how="left", validate="one_to_one")
)

#Verificação de anos com população ausente e aplicação de ffill para a utilizar a populacao do ano anterior caso necessário.
anos_pop_ausente = armas_anual.loc[armas_anual["populacao"].isna(), "ano"].tolist()
if anos_pop_ausente:
    print(f"Aviso metodológico: população UF indisponível em df_letalidade para {anos_pop_ausente}; aplicado ffill.")
    armas_anual["populacao"] = armas_anual["populacao"].ffill()

armas_anual["taxa_armas_populacao"] = taxa_por_100_mil(
    armas_anual["quantidade_arma_fogo_total"], armas_anual["populacao"]
)
armas_anual["taxa_fuzil_populacao"] = taxa_por_100_mil(
    armas_anual["quantidade_fuzil"], armas_anual["populacao"]
)

# Índice base 100 em 2007 para comparação em escala única
base_armas_2007 = armas_anual.loc[armas_anual["ano"] == 2007, "taxa_armas_populacao"].iloc[0]
base_fuzil_2007 = armas_anual.loc[armas_anual["ano"] == 2007, "taxa_fuzil_populacao"].iloc[0]
armas_anual["indice_taxa_armas"] = (armas_anual["taxa_armas_populacao"] / base_armas_2007) * 100
armas_anual["indice_taxa_fuzil"] = (armas_anual["taxa_fuzil_populacao"] / base_fuzil_2007) * 100

# Composição percentual do arsenal apreendido
armas_anual["prop_revolver_por_total"] = (
    armas_anual["quantidade_revolver"] / armas_anual["quantidade_arma_fogo_total"]
) * 100
armas_anual["prop_revolver_por_total"] = (
    armas_anual["quantidade_pistola"] / armas_anual["quantidade_arma_fogo_total"]
) * 100
armas_anual["prop_fuzil_por_total"] = (
    armas_anual["quantidade_fuzil"] / armas_anual["quantidade_arma_fogo_total"]
) * 100

# Resumo numérico para auditoria e redação dos achados
resumo_p1 = armas_anual[[
    "ano",
    "quantidade_arma_fogo_total",
    "taxa_armas_populacao",
    "quantidade_fuzil",
    "taxa_fuzil_populacao",
    "prop_revolver_por_total",
    "prop_revolver_por_total",
    "prop_fuzil_por_total",
]].round(2)

display(resumo_p1)
```

## Pergunta 2 — Concentração Territorial das Taxas de Letalidade Violenta e Feminicídio

> **Pergunta de domínio:** *Quais municípios ou Áreas Integradas (AISP) concentram as maiores taxas proporcionais de letalidade violenta e feminicídio?*

### 2.1 Transformação e Preparação

Para responder corretamente em termos **proporcionais** — corrigindo o viés populacional de contagens brutas — utilizamos `estatisticas_aisp` (construído na camada comum via `taxa_por_100_mil`) para a letalidade violenta e construímos um painel paralelo de feminicídio na mesma granularidade `(ano, id_aisp)` com `agregar_contagens_aisp_ano` aplicado a `df_fem_cisp`.

Três cuidados metodológicos são essenciais nesta etapa:

1. **Cobertura temporal do feminicídio:** os registros oficiais de `df_fem_cisp` iniciam-se em **novembro de 2016** (apenas 2 meses de coleta, totalizando 14 casos no estado, contra 68 em 2017). Para evitar distorção na taxa anual e na comparação temporal, restringimos o cálculo de taxas de feminicídio aos **anos completos a partir de 2017** que possuem denominador populacional em `populacao_por_ano_aisp`.

2. **Adequação do denominador populacional:** diferentemente da letalidade violenta, cuja taxa é calculada em relação à população total, a taxa de feminicídio deve considerar apenas a população feminina, por representar a população exposta ao fenômeno analisado. Diante da ausência de dados populacionais desagregados por sexo na base utilizada, adotamos a hipótese simplificadora de que a população feminina corresponde a **50% da população total de cada AISP**.

   Dessa forma, sendo $F$ a quantidade de feminicídios e $P$ a população total, temos:

   $$
   T_{\text{feminicídio}}
   = \frac{F}{P/2} \times 100.000
   = \frac{F}{P} \times 200.000
   $$

   Portanto, utilizamos o parâmetro `fator=200_000` na função `taxa_por_100_mil` para calcular as taxas de feminicídio por **100 mil mulheres**, mantendo o fator padrão de `100_000` para as taxas de letalidade violenta por **100 mil habitantes**. Essa aproximação constitui uma limitação metodológica, uma vez que a proporção feminina pode variar entre as AISPs e ao longo dos anos.

3. **Estabilidade estatística em eventos raros:** como o feminicídio apresenta contagens anuais reduzidas em cada AISP, taxas isoladas de um único ano sofrem alta variabilidade em territórios menores. Por isso, além do mapa interativo anual de letalidade violenta, calculamos a **taxa média anual ponderada por pessoas-ano** no período comum (`2017` até o último ano com população disponível):

   - Letalidade violenta:

   $$
   T_{\text{letalidade}}
   = \frac{\sum \text{ocorrências}}{\sum \text{população}}
   \times 100.000
   $$

   - Feminicídio:

   $$
   T_{\text{feminicídio}}
   = \frac{\sum \text{feminicídios}}{\sum \text{população}}
   \times 200.000
   $$

   Esse procedimento reduz a influência de oscilações anuais e permite uma comparação mais estável dos rankings espaciais de letalidade violenta e feminicídio, respeitando as respectivas populações de referência.

```python
# 2.1 Transformação e preparação — Pergunta 2
exigir_colunas(
    df_fem_cisp,
    ["ano", "id_aisp", "quantidade_morte_feminicidio", "quantidade_tentativa_feminicidio"],
    "df_fem_cisp",
)

# Painel anual de feminicídio por AISP (reutilizando funções da camada comum)
fem_aisp_ano = agregar_contagens_aisp_ano(
    df_fem_cisp,
    ["quantidade_morte_feminicidio", "quantidade_tentativa_feminicidio"],
)

ultimo_ano_pop = int(populacao_por_ano_aisp["ano"].max())
ano_inicio_fem_completo = 2017  # Escolha metodológica justificada acima

# Integração de feminicídio + população + metadados de dim_aisp
painel_fem_aisp = (
    populacao_por_ano_aisp.loc[
        populacao_por_ano_aisp["ano"].between(ano_inicio_fem_completo, ultimo_ano_pop)
    ]
    .merge(fem_aisp_ano, on=["ano", "id_aisp"], how="left", validate="one_to_one")
    .merge(
        dim_aisp[["id_aisp", "regiao_predominante", "rotulo_aisp"]],
        on="id_aisp",
        how="left",
        validate="many_to_one",
    )
)

#Preencher valores ausentes de feminicídio com 0, assumindo que a ausência de registro indica ausência de ocorrência.
painel_fem_aisp["quantidade_morte_feminicidio"] = (
    painel_fem_aisp["quantidade_morte_feminicidio"].fillna(0)
)

# Taxa por 100 mil mulheres, assumindo que a população feminina
# corresponde a 50% da população total
painel_fem_aisp["taxa_feminicidio"] = taxa_por_100_mil(
    painel_fem_aisp["quantidade_morte_feminicidio"],
    painel_fem_aisp["populacao"],
    fator=200_000,
)

# Enriquecer estatisticas_aisp com região predominante e rótulo para o mapa interativo
painel_let_aisp = (
    estatisticas_aisp.dropna(subset=["taxa_letalidade"])
    .merge(
        dim_aisp[["id_aisp", "regiao_predominante", "rotulo_aisp"]],
        on="id_aisp",
        how="left",
        validate="many_to_one",
    )
    .copy()
)

# A região recente da dimensão pode diferir da região histórica do código 1.
for painel in (painel_let_aisp, painel_fem_aisp):
    linhas = painel["id_aisp"].eq(1)
    nomes = painel.loc[linhas, "ano"].map(territorio_aisp1["nome_aisp"])
    painel.loc[linhas, "nome_aisp"] = nomes
    painel.loc[linhas, "rotulo_aisp"] = "AISP 01 — " + nomes
    painel.loc[linhas, "regiao_predominante"] = painel.loc[linhas, "ano"].map(
        df_cisp_mensal.loc[df_cisp_mensal["id_aisp"].eq(1)]
        .groupby("ano")["regiao"].agg(lambda s: s.mode().iloc[0])
    )

# Comparativo acumulado no período comum (2017–ultimo_ano_pop)
# com taxa ponderada por pessoas-ano
let_periodo_comum = (
    painel_let_aisp.loc[
        painel_let_aisp["ano"].between(ano_inicio_fem_completo, ultimo_ano_pop)
    ]
    .groupby(
        ["id_aisp", "nome_aisp", "rotulo_aisp", "regiao_predominante"],
        as_index=False,
    )
    .agg(
        total_letalidade=("quantidade_letalidade_violenta", "sum"),
        pessoas_ano=("populacao", "sum"),
    )
)

let_periodo_comum["taxa_media_letalidade"] = taxa_por_100_mil(
    let_periodo_comum["total_letalidade"],
    let_periodo_comum["pessoas_ano"],
)

fem_periodo_comum = (
    painel_fem_aisp.groupby("id_aisp", as_index=False)
    .agg(total_feminicidio=("quantidade_morte_feminicidio", "sum"))
)

comparativo_aisp_p2 = let_periodo_comum.merge(
    fem_periodo_comum,
    on="id_aisp",
    how="left",
    validate="one_to_one",
)

comparativo_aisp_p2["total_feminicidio"] = (
    comparativo_aisp_p2["total_feminicidio"].fillna(0)
)

# Taxa média por 100 mil mulheres, assumindo que a população feminina
# corresponde a 50% da população total
comparativo_aisp_p2["taxa_media_feminicidio"] = taxa_por_100_mil(
    comparativo_aisp_p2["total_feminicidio"],
    comparativo_aisp_p2["pessoas_ano"],
    fator=200_000,
)

# Correlação espacial entre taxas médias (Letalidade Violenta vs. Feminicídio)
r_espacial_let_fem = comparativo_aisp_p2["taxa_media_letalidade"].corr(
    comparativo_aisp_p2["taxa_media_feminicidio"]
)

# Série estadual anual completa (2017–2025)
# para comparação temporal sem 2016 incompleto
fem_anual_uf = (
    df_fem_cisp.groupby("ano", as_index=False)["quantidade_morte_feminicidio"]
    .sum()
)

let_anual_uf = (
    mortes_aisp_ano.groupby("ano", as_index=False)["quantidade_letalidade_violenta"]
    .sum()
)

serie_let_fem_uf = (
    fem_anual_uf.merge(let_anual_uf, on="ano", validate="one_to_one")
    .query("2017 <= ano <= 2025")
    .sort_values("ano")
)

r_temporal_let_fem_2017_2025 = serie_let_fem_uf["quantidade_morte_feminicidio"].corr(
    serie_let_fem_uf["quantidade_letalidade_violenta"]
)

print(f"Período comum de taxas por AISP (com anos completos de feminicídio): {ano_inicio_fem_completo}–{ultimo_ano_pop}")
print(f"Correlação espacial entre taxa média de letalidade e feminicídio por AISP (r): {r_espacial_let_fem:.4f}")
print(f"Correlação temporal estadual Letalidade vs. Feminicídio (2017–2025, r): {r_temporal_let_fem_2017_2025:.4f}")

print("\nTop 10 AISPs — Taxa média anual de Letalidade Violenta (por 100 mil habitantes):")
display(
    comparativo_aisp_p2.nlargest(10, "taxa_media_letalidade")[
        [
            "rotulo_aisp",
            "regiao_predominante",
            "taxa_media_letalidade",
            "total_letalidade",
            "taxa_media_feminicidio",
        ]
    ].round(2)
)

print("\nTop 10 AISPs — Taxa média anual de Feminicídio (por 100 mil mulheres):")
display(
    comparativo_aisp_p2.nlargest(10, "taxa_media_feminicidio")[
        [
            "rotulo_aisp",
            "regiao_predominante",
            "taxa_media_feminicidio",
            "total_feminicidio",
            "taxa_media_letalidade",
        ]
    ].round(2)
)
```

## Pergunta 3 — Sazonalidade Mensal da Letalidade Violenta por Região de Policiamento

> **Pergunta de domínio:** *Como se comporta a sazonalidade da letalidade violenta ao longo dos meses do ano nas diferentes regiões de policiamento do estado?*

### 3.1 Transformação e Preparação

A partir de `df_cisp_mensal` (já padronizado na camada comum), filtramos os anos completos da série histórica (**2003 a 2024**) e agregamos `quantidade_letalidade_violenta` por `regiao` (`Capital`, `Baixada Fluminense`, `Grande Niterói` e `Interior`) e `mes`, além de construir a linha de referência estadual (`Estado (RJ)`).

**Ajuste de calendário e Índice Sazonal (Base 100):** meses possuem durações diferentes (fevereiro tem em média 28,25 dias, contra 31 dias em janeiro e março — diferença próxima de 10%). Para evitar que fevereiro pareça artificialmente menos violento apenas por ser mais curto, convertemos a soma mensal em **média diária de vítimas** dividindo pelo número médio de dias de cada mês e, em seguida, calculamos o **Índice Sazonal (Base 100)** em relação à média diária anual de cada região:
- Valores **acima de 100** indicam meses com intensidade diária acima da média daquela região;
- Valores **abaixo de 100** indicam meses relativamente mais brandos.

```python
# 3.1 Transformação e preparação — Pergunta 3
exigir_colunas(df_cisp_mensal, ["ano", "mes", "regiao", "quantidade_letalidade_violenta"], "df_cisp_mensal")

regioes_ordem_p3 = ["Capital", "Baixada Fluminense", "Grande Niterói", "Interior"]
meses_rotulos = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"]
dias_por_mes = pd.Series(
    [31, 28.25, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31],
    index=range(1, 13),
    name="dias_mes",
)

base_p3 = df_cisp_mensal.loc[
    df_cisp_mensal["ano"].between(2003, 2024)
    & df_cisp_mensal["regiao"].isin(regioes_ordem_p3)
].copy()

# Soma mensal por região (2003–2024) e ajuste por dias médios do mês
sazonal_reg = (
    base_p3.groupby(["regiao", "mes"])["quantidade_letalidade_violenta"]
    .sum()
    .unstack("mes")
    .reindex(index=regioes_ordem_p3, columns=range(1, 13))
)
sazonal_estado = (
    base_p3.groupby("mes")["quantidade_letalidade_violenta"]
    .sum()
    .reindex(range(1, 13))
    .rename("Estado (RJ)")
    .to_frame()
    .T
)
sazonal_bruto = pd.concat([sazonal_reg, sazonal_estado], axis=0)

# Taxa diária relativa e Índice Sazonal Base 100
intensidade_diaria = sazonal_bruto.div(dias_por_mes, axis=1)
indice_sazonal = intensidade_diaria.div(intensidade_diaria.mean(axis=1), axis=0).mul(100)
indice_sazonal.columns = meses_rotulos

# Resumo de picos, vales e amplitude sazonal por região
resumo_p3 = pd.DataFrame({
    "mes_pico": indice_sazonal.idxmax(axis=1),
    "indice_pico": indice_sazonal.max(axis=1).round(2),
    "mes_vale": indice_sazonal.idxmin(axis=1),
    "indice_vale": indice_sazonal.min(axis=1).round(2),
    "amplitude_pts": (indice_sazonal.max(axis=1) - indice_sazonal.min(axis=1)).round(2),
})

print("Índice Sazonal Base 100 (ajustado pela duração dos meses, 2003–2024):")
display(indice_sazonal.round(1))
print("\nResumo de picos, vales e amplitude sazonal:")
display(resumo_p3)
```

## Pergunta 5 — Clusters Territoriais: Intervenção Policial e Apreensão de Armamento Pesado

> **Pergunta de domínio:** *O mapeamento territorial revela agrupamentos (clusters) onde altas taxas de mortes por intervenção policial coexistem de forma consistente com volumes elevados de apreensão de armamento pesado?*

### 5.1 Transformação e Preparação

Para testar a coexistência territorial entre letalidade policial e armamento de guerra sem distorções pontuais de um único ano, consolidamos o período recente com cobertura populacional completa (**2017–2022**, alinhado à Pergunta 2) por `id_aisp`, reutilizando `agregar_contagens_aisp_ano`, `taxa_por_100_mil`, `populacao_por_ano_aisp` e `dim_aisp`:
- **Eixo Y (Letalidade Policial):** taxa média anual de mortes por intervenção de agente do Estado por 100 mil habitantes (`quantidade_homicidio_intervencao_policial` dividida por pessoas-ano);
- **Eixo X (Armamento Pesado):** volume acumulado de fuzis apreendidos (`quantidade_fuzil`), acompanhado da taxa proporcional de fuzis (`taxa_fuzil_populacao`) como verificação de robustez.

Classificamos as 41 AISPs em **quatro quadrantes (clusters)** definidos pelas **medianas** de cada variável (critério robusto a valores extremos), destacando o **Cluster 1 (Foco Crítico: Alta MDIP + Alto Volume de Fuzis)**.

```python
# 5.1 Transformação e preparação — Pergunta 5
exigir_colunas(df_cisp_mensal, ["ano", "id_aisp", "quantidade_homicidio_intervencao_policial"], "df_cisp_mensal")
exigir_colunas(df_armas, ["ano", "id_aisp", "quantidade_fuzil"], "df_armas")

ano_ini_p5, ano_fim_p5 = 2017, int(populacao_por_ano_aisp["ano"].max())

mdip_aisp_ano = agregar_contagens_aisp_ano(
    df_cisp_mensal.loc[df_cisp_mensal["ano"].between(ano_ini_p5, ano_fim_p5)],
    ["quantidade_homicidio_intervencao_policial"],
)
fuzil_aisp_ano = agregar_contagens_aisp_ano(
    df_armas.loc[df_armas["ano"].between(ano_ini_p5, ano_fim_p5)],
    ["quantidade_fuzil"],
)

pop_p5 = (
    populacao_por_ano_aisp.loc[populacao_por_ano_aisp["ano"].between(ano_ini_p5, ano_fim_p5)]
    .groupby("id_aisp", as_index=False)
    .agg(pessoas_ano=("populacao", "sum"), pop_media=("populacao", "mean"))
)

painel_p5 = (
    pop_p5
    .merge(
        mdip_aisp_ano.groupby("id_aisp", as_index=False)["quantidade_homicidio_intervencao_policial"].sum(),
        on="id_aisp", how="left", validate="one_to_one",
    )
    .merge(
        fuzil_aisp_ano.groupby("id_aisp", as_index=False)["quantidade_fuzil"].sum(),
        on="id_aisp", how="left", validate="one_to_one",
    )
    .merge(
        dim_aisp[["id_aisp", "nome_aisp", "rotulo_aisp", "regiao_predominante"]],
        on="id_aisp", how="left", validate="one_to_one",
    )
)

painel_p5["quantidade_homicidio_intervencao_policial"] = painel_p5["quantidade_homicidio_intervencao_policial"].fillna(0)
painel_p5["quantidade_fuzil"] = painel_p5["quantidade_fuzil"].fillna(0)

painel_p5["taxa_mdip_100k"] = taxa_por_100_mil(
    painel_p5["quantidade_homicidio_intervencao_policial"], painel_p5["pessoas_ano"]
)
painel_p5["taxa_fuzil_populacao"] = taxa_por_100_mil(
    painel_p5["quantidade_fuzil"], painel_p5["pessoas_ano"]
)

# Linhas de corte (medianas) para definição dos 4 quadrantes/clusters
mediana_fuzil = float(painel_p5["quantidade_fuzil"].median())
mediana_mdip = float(painel_p5["taxa_mdip_100k"].median())


def classificar_cluster(row):
    alto_fuzil = row["quantidade_fuzil"] >= mediana_fuzil
    alta_mdip = row["taxa_mdip_100k"] >= mediana_mdip
    if alto_fuzil and alta_mdip:
        return "Cluster 1: Alta MDIP + Alto Fuzil"
    if (not alto_fuzil) and alta_mdip:
        return "Cluster 2: Alta MDIP + Baixo Fuzil"
    if alto_fuzil and (not alta_mdip):
        return "Cluster 3: Baixa MDIP + Alto Fuzil"
    return "Cluster 4: Baixa MDIP + Baixo Fuzil"


painel_p5["cluster"] = painel_p5.apply(classificar_cluster, axis=1)

# Correlações (Pearson e Spearman)
r_pearson_vol = painel_p5["quantidade_fuzil"].corr(painel_p5["taxa_mdip_100k"], method="pearson")
r_spearman_vol = painel_p5["quantidade_fuzil"].corr(painel_p5["taxa_mdip_100k"], method="spearman")
r_spearman_taxa = painel_p5["taxa_fuzil_populacao"].corr(painel_p5["taxa_mdip_100k"], method="spearman")

print(f"Período consolidado P5: {ano_ini_p5}–{ano_fim_p5} ({len(painel_p5)} AISPs)")
print(f"Medianas de corte: Fuzis acumulados = {mediana_fuzil:.1f} | Taxa MDIP = {mediana_mdip:.2f} / 100 mil hab.")
print(f"Correlação Volume de Fuzis vs. Taxa MDIP: Pearson r = {r_pearson_vol:.4f} | Spearman rho = {r_spearman_vol:.4f}")
print(f"Correlação Taxa de Fuzis vs. Taxa MDIP (robustez): Spearman rho = {r_spearman_taxa:.4f}")
print("\nDistribuição das AISPs por cluster:")
display(painel_p5["cluster"].value_counts().rename_axis("cluster").reset_index(name="qtd_aisps"))

print("\nAISPs no Cluster 1 (Foco Crítico — ordenadas por Taxa MDIP):")
display(
    painel_p5.loc[painel_p5["cluster"] == "Cluster 1: Alta MDIP + Alto Fuzil"]
    .sort_values("taxa_mdip_100k", ascending=False)[
        ["rotulo_aisp", "regiao_predominante", "taxa_mdip_100k", "quantidade_homicidio_intervencao_policial", "quantidade_fuzil", "taxa_fuzil_populacao"]
    ]
    .round(2)
)
```

## Pergunta 6 — Peso das Intervenções Policiais na Letalidade dos Municípios Mais Populosos

> **Pergunta de domínio:** *A variação na proporção de mortes por intervenção policial afeta a tendência geral de letalidade violenta a longo prazo nos municípios mais populosos?*

### 6.1 Transformação e Preparação

Integramos `df_mun_mensal` (`quantidade_letalidade_violenta`, `quantidade_homicidio_intervencao_policial`, `quantidade_registro_ocorrencia`) com `df_taxa_mun_mensal` (`taxa_letalidade_violenta`, `taxa_registro_ocorrencia`) para o período completo de **2003 a 2024** (ou último ano disponível nas duas bases):
1. Estimamos a população média municipal diretamente pela identidade oficial $\text{pop} = \frac{\text{quantidade\_registro\_ocorrencia}}{\text{taxa\_registro\_ocorrencia}} \times 100.000$ (evitando denominadores nulos) para selecionar os **6 municípios mais populosos do Rio de Janeiro**;
2. Agregamos anualmente, por município, a **taxa anual de letalidade violenta por 100 mil habitantes** (soma das taxas mensais oficiais em `df_taxa_mun_mensal`) e a **participação percentual (%) das mortes por intervenção policial** na letalidade violenta (`prop_mdip_pct = MDIP / Letalidade Violenta × 100`);
3. Calculamos a correlação temporal de Pearson ($r$) entre `prop_mdip_pct` e `taxa_letalidade_anual` em cada município — tanto em nível quanto em variação anual ($\Delta$) — para avaliar se picos de participação policial acompanham alta ou queda da letalidade geral.

```python
# 6.1 Transformação e preparação — Pergunta 6
exigir_colunas(
    df_mun_mensal,
    ["ano", "mes", "id_municipio", "quantidade_letalidade_violenta", "quantidade_homicidio_intervencao_policial", "quantidade_registro_ocorrencia"],
    "df_mun_mensal",
)
exigir_colunas(
    df_taxa_mun_mensal,
    ["ano", "mes", "id_municipio", "taxa_letalidade_violenta", "taxa_registro_ocorrencia"],
    "df_taxa_mun_mensal",
)

# Nomes oficiais dos principais municípios fluminenses (códigos IBGE)
NOMES_MUNICIPIOS_RJ = {
    3304557: "Rio de Janeiro",
    3304904: "São Gonçalo",
    3301702: "Duque de Caxias",
    3303500: "Nova Iguaçu",
    3303302: "Niterói",
    3300456: "Belford Roxo",
    3301009: "Campos dos Goytacazes",
    3305109: "São João de Meriti",
    3303906: "Petrópolis",
    3306305: "Volta Redonda",
}

base_mun_p6 = df_mun_mensal[
    ["ano", "mes", "id_municipio", "quantidade_letalidade_violenta", "quantidade_homicidio_intervencao_policial", "quantidade_registro_ocorrencia"]
].merge(
    df_taxa_mun_mensal[["ano", "mes", "id_municipio", "taxa_letalidade_violenta", "taxa_registro_ocorrencia"]],
    on=["ano", "mes", "id_municipio"],
    how="inner",
    validate="one_to_one",
)

# Restringir a anos completos (12 meses disponíveis)
meses_por_ano_mun = base_mun_p6.groupby("ano")["mes"].nunique()
anos_completos_p6 = meses_por_ano_mun.loc[meses_por_ano_mun == 12].index
base_mun_p6 = base_mun_p6.loc[base_mun_p6["ano"].isin(anos_completos_p6)].copy()

# Estimativa populacional para ranquear os 6 municípios mais populosos no período recente
taxa_ro_valida = base_mun_p6["taxa_registro_ocorrencia"].where(base_mun_p6["taxa_registro_ocorrencia"] > 0)
base_mun_p6["pop_estimada"] = (base_mun_p6["quantidade_registro_ocorrencia"] / taxa_ro_valida) * FATOR_TAXA

ano_ref_pop = int(base_mun_p6["ano"].max())
top6_ids = (
    base_mun_p6.loc[base_mun_p6["ano"] >= ano_ref_pop - 2]
    .groupby("id_municipio")["pop_estimada"]
    .mean()
    .nlargest(6)
    .index.tolist()
)

# Painel anual dos Top 6 municípios
painel_p6 = (
    base_mun_p6.loc[base_mun_p6["id_municipio"].isin(top6_ids)]
    .groupby(["ano", "id_municipio"], as_index=False)
    .agg(
        letalidade=("quantidade_letalidade_violenta", "sum"),
        mdip=("quantidade_homicidio_intervencao_policial", "sum"),
        taxa_letalidade_anual=("taxa_letalidade_violenta", "sum"),
        pop_media=("pop_estimada", "mean"),
    )
)
painel_p6["nome_municipio"] = (
    painel_p6["id_municipio"].map(NOMES_MUNICIPIOS_RJ).fillna(painel_p6["id_municipio"].astype(str))
)
painel_p6["prop_mdip_pct"] = np.where(
    painel_p6["letalidade"] > 0,
    (painel_p6["mdip"] / painel_p6["letalidade"]) * 100,
    np.nan,
)
painel_p6 = painel_p6.sort_values(["id_municipio", "ano"]).reset_index(drop=True)

# Resumo estatístico por município (tendência + correlação em nível e primeira diferença)
ordem_mun_p6 = [NOMES_MUNICIPIOS_RJ.get(i, str(i)) for i in top6_ids]
linhas_resumo_p6 = []
for mun in ordem_mun_p6:
    sub = painel_p6.loc[painel_p6["nome_municipio"] == mun].sort_values("ano")
    r_nivel = sub["prop_mdip_pct"].corr(sub["taxa_letalidade_anual"])
    r_diff = sub["prop_mdip_pct"].diff().corr(sub["taxa_letalidade_anual"].diff())
    linhas_resumo_p6.append({
        "municipio": mun,
        "ano_ini": int(sub["ano"].min()),
        "ano_fim": int(sub["ano"].max()),
        "taxa_ini": sub["taxa_letalidade_anual"].iloc[0],
        "taxa_fim": sub["taxa_letalidade_anual"].iloc[-1],
        "prop_mdip_ini_pct": sub["prop_mdip_pct"].iloc[0],
        "prop_mdip_fim_pct": sub["prop_mdip_pct"].iloc[-1],
        "prop_mdip_max_pct": sub["prop_mdip_pct"].max(),
        "ano_pico_mdip": int(sub.loc[sub["prop_mdip_pct"].idxmax(), "ano"]),
        "r_nivel": r_nivel,
        "r_variacao_anual": r_diff,
    })

resumo_p6 = pd.DataFrame(linhas_resumo_p6).round(2)
print(f"Período completo de municípios analisado: {int(painel_p6['ano'].min())}–{int(painel_p6['ano'].max())}")
display(resumo_p6)
```

## Pergunta 7 — Correlação Temporal entre Policiais Mortos em Serviço e Letalidade Violenta Subsequente

> **Pergunta de domínio:** *Existe correlação temporal observável (como picos e vales simultâneos) entre a quantidade de policiais mortos em serviço e a escalada da letalidade violenta geral nos meses subsequentes?*

### 7.1 Transformação e Preparação

Integramos mensalmente `df_policial` (`evolucao_policial_morto_servico_mensal`), somando militares e civis mortos em serviço (`policiais_mortos_servico = quantidade_policial_militar_morto_servico + quantidade_policial_civil_morto_servico`), à série estadual mensal `df_evol_uf` (`quantidade_letalidade_violenta` e `quantidade_homicidio_intervencao_policial`).

**Rigor estatístico na análise de defasagens (lags 0, 1, 2 e 3 meses):**
Em séries temporais longas, duas variáveis que compartilham uma tendência secular comum (por exemplo, ambas em queda entre 2003 e 2024) tendem a apresentar correlação positiva elevada em nível mesmo sem resposta dinâmica imediata de um mês para o outro. Para distinguir **tendência compartilhada** de **escalada no mês subsequente**, calculamos a correlação cruzada $\text{Corr}(X_t, Y_{t+k})$ para $k \in \{0, 1, 2, 3\}$ meses sob dois regimes:
1. **Em nível** (série mensal observada);
2. **Em primeira diferença mensal** ($\Delta X_t = X_t - X_{t-1}$ vs. $\Delta Y_{t+k} = Y_{t+k} - Y_{t+k-1}$), que remove a tendência de longo prazo e isola choques de curto prazo, reportando também o limite de significância aproximado ($\pm 1{,}96 / \sqrt{n}$).

```python
# 7.1 Transformação e preparação — Pergunta 7
exigir_colunas(
    df_policial,
    ["ano", "mes", "quantidade_policial_militar_morto_servico", "quantidade_policial_civil_morto_servico"],
    "df_policial",
)
exigir_colunas(
    df_evol_uf,
    ["ano", "mes", "quantidade_letalidade_violenta", "quantidade_homicidio_intervencao_policial"],
    "df_evol_uf",
)

pol_mensal = (
    df_policial.groupby(["ano", "mes"], as_index=False)[
        ["quantidade_policial_militar_morto_servico", "quantidade_policial_civil_morto_servico"]
    ]
    .sum()
)
pol_mensal["policiais_mortos_servico"] = (
    pol_mensal["quantidade_policial_militar_morto_servico"]
    + pol_mensal["quantidade_policial_civil_morto_servico"]
)

serie_p7 = (
    pol_mensal.merge(
        df_evol_uf[["ano", "mes", "quantidade_letalidade_violenta", "quantidade_homicidio_intervencao_policial"]],
        on=["ano", "mes"],
        how="inner",
        validate="one_to_one",
    )
    .sort_values(["ano", "mes"])
    .reset_index(drop=True)
)
serie_p7["data"] = pd.to_datetime({"year": serie_p7["ano"], "month": serie_p7["mes"], "day": 1})

# Médias móveis de 6 meses para leitura visual de ciclos
serie_p7["mm6_policiais"] = serie_p7["policiais_mortos_servico"].rolling(6, center=True).mean()
serie_p7["mm6_letalidade"] = serie_p7["quantidade_letalidade_violenta"].rolling(6, center=True).mean()

# Primeiras diferenças mensais (remove tendência secular)
serie_p7["d_policiais"] = serie_p7["policiais_mortos_servico"].diff()
serie_p7["d_letalidade"] = serie_p7["quantidade_letalidade_violenta"].diff()
serie_p7["d_mdip"] = serie_p7["quantidade_homicidio_intervencao_policial"].diff()

# Correlações cruzadas para lags k = 0, 1, 2, 3 meses: Corr(X_t, Y_{t+k})
lags = [0, 1, 2, 3]
resultados_lag = []
for k in lags:
    r_let_nivel = serie_p7["policiais_mortos_servico"].corr(serie_p7["quantidade_letalidade_violenta"].shift(-k))
    r_let_diff = serie_p7["d_policiais"].corr(serie_p7["d_letalidade"].shift(-k))
    r_mdip_nivel = serie_p7["policiais_mortos_servico"].corr(serie_p7["quantidade_homicidio_intervencao_policial"].shift(-k))
    r_mdip_diff = serie_p7["d_policiais"].corr(serie_p7["d_mdip"].shift(-k))
    resultados_lag.append({
        "lag_meses": k,
        "r_letalidade_nivel": r_let_nivel,
        "r_letalidade_diff": r_let_diff,
        "r_mdip_nivel": r_mdip_nivel,
        "r_mdip_diff": r_mdip_diff,
    })

resumo_lags_p7 = pd.DataFrame(resultados_lag).round(4)
n_meses_p7 = len(serie_p7)
limiar_sig_95 = 1.96 / np.sqrt(n_meses_p7)

print(
    f"Série mensal integrada P7: {serie_p7['data'].min().strftime('%Y-%m')} a {serie_p7['data'].max().strftime('%Y-%m')} "
    f"(n = {n_meses_p7} meses | limiar ~95% para ruído branco: ±{limiar_sig_95:.3f})"
)
print(
    f"Total de policiais mortos em serviço no período: {int(serie_p7['policiais_mortos_servico'].sum())} "
    f"(pico mensal: {int(serie_p7['policiais_mortos_servico'].max())} em "
    f"{serie_p7.loc[serie_p7['policiais_mortos_servico'].idxmax(), 'data'].strftime('%Y-%m')})"
)
display(resumo_lags_p7)
```

### 7.2 Visualização

### 7.3 Análise e Achados

1. **Co-movimento em nível reflete ciclos históricos compartilhados (2003–2026):** Ao longo dos **284 meses** integrados (janeiro de 2003 a agosto de 2026), registraram-se **596 policiais mortos em serviço** (pico mensal de **10** em outubro de 2004). Na série em nível, a correlação com a letalidade violenta estadual mantém-se moderada e praticamente estável em todas as defasagens — **lag 0: $r = +0{,}441$**; **lag 1: $r = +0{,}415$**; **lag 2: $r = +0{,}389$**; **lag 3: $r = +0{,}447$** (e, contra MDIP em nível, entre $+0{,}107$ e $+0{,}163$). Essa estabilidade entre lags 0 e 3 é a assinatura típica de uma **tendência secular comum** (ambas as séries partiram de patamares elevados nos anos 2000 e declinaram ao longo de duas décadas), e não de uma reação específica no mês imediatamente posterior.
2. **Ausência de escalada sistemática em primeiras diferenças (choques de curto prazo):** Ao remover a tendência secular por meio de primeiras diferenças mensais ($\Delta X_t$ vs. $\Delta Y_{t+k}$), a correlação desaparece no mês contemporâneo e no mês seguinte: para a letalidade violenta total, **lag 0 = $+0{,}059$** e **lag 1 = $+0{,}005$** (muito abaixo da banda de significância aproximada de $\pm 0{,}116$); contra mortes por intervenção policial especificamente, **todos os quatro lags permanecem dentro da faixa de ruído branco** (**lag 0 = $+0{,}026$**, **lag 1 = $+0{,}015$**, **lag 2 = $-0{,}093$**, **lag 3 = $+0{,}023$**). As oscilações modestas e de sinais opostos em lag 2 ($-0{,}162$) e lag 3 ($+0{,}147$) na letalidade total sugerem reversão à média/ruído sazonal, não uma escalada sustentada.
3. **Interpretação estatística rigorosa (sem inferência causal):** Os dados estaduais mensais **não corroboram** uma relação causal direta de que meses com mais policiais mortos em serviço provoquem automaticamente uma escalada generalizada da letalidade violenta ou da MDIP nos meses $t+1$ a $t+3$. Eventuais respostas operacionais após mortes de agentes tendem a ocorrer em escalas **microlocais** (CISP/AISP) e janelas de **dias ou semanas**, diluindo-se na agregação mensal estadual.

# Conclusão

## Síntese Integrada dos Achados

A análise longitudinal e georreferenciada das bases oficiais do ISP-RJ demonstra que a segurança pública fluminense não pode ser compreendida por um único indicador agregado. Embora a taxa estadual de letalidade violenta tenha recuado expressivamente no longo prazo (de 60,3 em 1991 e 54,3 em 2003 para 25,5 por 100 mil hab. em 2022) e o volume total de armas apreendidas tenha caído 44,4% entre 2007 e 2024, **a violência letal transformou-se qualitativamente e redistribuiu-se no território**:

- **Maior poder de fogo em menos armas (P1):** os fuzis mais que triplicaram em volume (+242,1%, de 214 para 732) e sextuplicaram sua participação no arsenal apreendido (1,93% → 11,90%), acompanhados pela predominância das pistolas (50,96% em 2024) sobre os revólveres (25,51%).
- **Convergência regional e interiorização relativa (P4):** a queda acentuada na Capital (−66,7% entre 2003 e 2022) e em Grande Niterói (−52,7%) contrastou com a maior rigidez no Interior (−24,6%), invertendo o ordenamento histórico frente à Capital em 2022 (Interior 26,0 vs. Capital 19,4/100 mil), enquanto a Baixada Fluminense manteve o maior patamar proporcional do estado (35,9/100 mil).
- **Fenômenos espacialmente distintos (P2 e P5):** a letalidade violenta e a letalidade policial concentram-se em um corredor metropolitano contínuo fortemente associado à apreensão de fuzis (15 das 39 AISPs no Cluster 1; Spearman $\rho = 0{,}834$), ao passo que o **feminicídio** exibe geografia própria (6 das 10 maiores taxas no Interior; $r_{\text{espacial}} = 0{,}354$) e trajetória temporal ascendente em 2017–2025 ($r = -0{,}869$ frente à letalidade geral).
- **Sazonalidade, composição municipal e dinâmica de curto prazo (P3, P6 e P7):** todas as regiões compartilham picos de intensidade diária em fevereiro–março e vales em julho–agosto (amplitude estadual de 22,2 pontos); nos grandes municípios metropolitanos (2014–2024), a participação de MDIP chegou a 55,7% em Niterói e 48,7% em São Gonçalo mesmo em contexto de queda da letalidade geral; por fim, a associação entre policiais mortos em serviço e letalidade estadual ao longo de 284 meses decorre de tendência histórica compartilhada, sem evidência de escalada sistemática no mês seguinte em primeiras diferenças ($r_{\text{lag 1}} = +0{,}005$).

## Recomendações Baseadas em Dados

1. **Focalização territorial diferenciada por tipo de violência:** políticas de controle de armamento pesado e redução de confrontos letais devem priorizar o **corredor crítico do Cluster 1** (eixo Zona Norte–Baixada–São Gonçalo e Angra dos Reis), com foco em inteligência de interceptação de fuzis e protocolos de redução de letalidade policial.
2. **Interiorização das políticas de proteção à mulher:** dado que o feminicídio concentra suas maiores taxas proporcionais no Interior (Noroeste e Norte Fluminense), a expansão de Delegacias Especializadas de Atendimento à Mulher (DEAMs), patrulhas Maria da Penha e redes de acolhimento precisa descentralizar-se além da Região Metropolitana.
3. **Planejamento operacional sensível à sazonalidade e monitoramento microlocal:** a alocação preventiva de efetivo pode antecipar a elevação consistente da intensidade diária no primeiro quadrimestre (`Jan–Abr`, especialmente `Fev–Mar`), enquanto o acompanhamento de eventos críticos envolvendo agentes de segurança deve ser feito em escala diária e por CISP/AISP, onde choques locais são detectáveis.

## Limitações Metodológicas

1. **Cobertura temporal das séries populacionais:** a base `PopulacaoEvolucaoMensalCisp` e a população estadual de `df_letalidade` encerram-se em **2022**. Assim, análises de taxas por AISP limitam-se a `2003–2022` (e `2017–2022` no cruzamento com feminicídio), e as taxas estaduais de armas em 2023–2024 dependem de propagação (`ffill`) da população de 2022.
2. **Ausência de vigência temporal na tabela CISP–AISP–RISP:** o arquivo `relacao_cisp_aisp_risp` reflete a divisão territorial consolidada mais recente, e não contém o histórico das fronteiras. As taxas usam a composição por CISP observada em cada mês. Comparações entre anos podem abranger territórios diferentes mesmo quando o código coincide.
3. **Natureza administrativa dos registros policiais:** todas as métricas derivam de Registros de Ocorrência (ROs). Mudanças de tipificação legal (como a consolidação da Lei do Feminicídio nº 13.104/2015, com coleta iniciada em novembro de 2016), variações na subnotificação e alterações na intensidade do policiamento afetam as séries observadas, impondo interpretação descritiva e exploratória — nunca causalidade isolada.
