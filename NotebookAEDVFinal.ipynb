## Introdução

O estado do Rio de Janeiro vivencia, há décadas, um dos cenários de segurança pública mais complexos e militarizados da América Latina. O presente artigo investiga a arquitetura espaço-temporal da violência letal fluminense a partir da integração de 16 bases abertas do Instituto de Segurança Pública (ISP-RJ) e de três malhas geográficas oficiais (CISP, AISP e RISP), cobrindo séries longitudinais entre 1991 e 2026. Estruturado em torno de sete perguntas de domínio em progressão lógica — da descrição macro à análise espaço-temporal e de defasagens —, o estudo examina desde a transformação tecnológica nos arsenais apreendidos até a concentração territorial da letalidade policial, do feminicídio e dos ciclos sazonais.

### Síntese dos Principais Achados
- **Militarização do arsenal (P1):** Entre 2007 e 2024, o total anual de armas de fogo apreendidas caiu **44,4%** (de 11.062 para 6.150; taxa de 71,47 para 35,02/100 mil hab.), enquanto as apreensões de **fuzis mais que triplicaram** (+242,1%, de 214 para 732; fatia saltando de 1,93% para 11,90%), com substituição estrutural de revólveres por pistolas.
- **Geografia distinta entre letalidade geral e feminicídio (P2):** Normalizadas pela população (2017–2022), as maiores taxas de letalidade violenta concentram-se na Baixada Fluminense e polos litorâneos (lideradas por AISP 24 — Itaguaí/Seropédica, com 58,13/100 mil), enquanto **6 das 10 maiores taxas de feminicídio estão no Interior** (correlação espacial baixa, $r = 0{,}354$, e divergência temporal acentuada em 2017–2025, $r = -0{,}869$).
- **Sazonalidade sincronizada (P3):** Ajustada pela duração dos meses (2003–2024), a letalidade violenta apresenta picos consistentes em **fevereiro–março** (índice 112,0 no estado) e vales no inverno (**julho–agosto**, índice 89,8), com maior amplitude no Interior (25,1 pontos).
- **Queda metropolitana e convergência regional (P4):** De 2003 a 2022, a taxa caiu **66,7%** na Capital (58,3 → 19,4/100 mil) e **49,0%** na Baixada Fluminense (70,4 → 35,9/100 mil), frente a **−24,6%** no Interior (34,5 → 26,0/100 mil), reduzindo a razão Máx/Mín regional de 2,04× para 1,85×.
- **Clusters de confronto armado (P5):** Há forte associação espacial entre volume de fuzis apreendidos e taxa de mortes por intervenção policial ($r = 0{,}757$; Spearman $\rho = 0{,}834$), com **15 das 39 AISPs** formando um corredor crítico contínuo na Região Metropolitana (mais Angra dos Reis).
- **Peso da intervenção policial nos grandes municípios (P6):** Entre 2014 e 2024, nos 6 municípios mais populosos, a participação da MDIP chegou a **55,7%** em Niterói (2023) e **48,7%** em São Gonçalo (2021), permanecendo residual em Campos dos Goytacazes (máx. 3,6%); enquanto variações anuais caminham juntas ($r_{\Delta}$ entre $+0{,}22$ e $+0{,}61$), no longo prazo a fatia de MDIP cresceu mesmo com queda da letalidade geral em 4 dos 6 municípios.
- **Vitimização policial e letalidade subsequente (P7):** Ao longo de 284 meses (2003–2026; 596 policiais mortos em serviço), a correlação positiva em nível ($r \approx 0{,}39$ a $0{,}45$) reflete tendência histórica compartilhada; ao remover a tendência em primeiras diferenças, o efeito no mês seguinte desaparece ($r_{\text{lag 1}} = +0{,}005$ para letalidade total e $+0{,}015$ para MDIP), afastando causalidade mecânica no agregado mensal estadual.

## Contextualização e Conceitos-Chave

1. **Sobre os dados:** Os dados têm origem no Instituto de Segurança Pública do Estado do Rio de Janeiro (ISP-RJ) e estão estruturados em um conjunto de bases abertas de frequências mensal e anual. A coleta e a agregação originais são feitas a partir dos registros de ocorrência das delegacias do estado. As variáveis mais relevantes abrangem recortes temporais e geográficos, tipificações de armas apreendidas (com destaque para fuzis) e contagens e taxas (por 100 mil habitantes) de crimes contra a vida, além da tabela de relação CISP–AISP–RISP para cruzamentos espaciais entre delegacia, município e estado.

2. **Siglas territoriais:**
   - **CISP:** A Circunscrição Integrada de Segurança Pública é a área territorial que coincide com a jurisdição de uma delegacia distrital da Polícia Civil (SEPOL) e a subárea correspondente de uma companhia da Polícia Militar (SEPM).
   - **AISP:** A Área Integrada de Segurança Pública é uma divisão territorial que alinha a atuação da Polícia Militar (PMERJ) e da Polícia Civil (PCERJ) na mesma região para combater a criminalidade de forma conjunta.
   - **RISP:** A Região Integrada de Segurança Pública une os comandos intermediários das polícias, ou seja, os Departamentos de Polícia de Área (DPA) da Polícia Civil e os Comandos de Policiamento de Área (CPA) da Polícia Militar.

3. **Definição oficial de Letalidade Violenta (ISP-RJ):**
   - O indicador agregado de **Letalidade Violenta** utilizado oficialmente pelo ISP-RJ soma quatro tipificações de crimes contra a vida: **homicídio doloso**, **latrocínio** (roubo seguido de morte), **lesão corporal seguida de morte** e **morte por intervenção de agente do Estado** (autos de resistência / intervenção policial).

```python
def exigir_colunas(base: pd.DataFrame, colunas: list[str], nome_base: str) -> None:
    '''Valida a presença de colunas obrigatórias em um DataFrame.

    Args:
        base: DataFrame a ser validado.
        colunas: Lista com os nomes das colunas exigidas.
        nome_base: Identificador textual da tabela exibido em caso de erro.

    Returns:
        None.

    Raises:
        KeyError: Se uma ou mais colunas especificadas não existirem em ``base``.
    '''
    faltantes = set(colunas).difference(base.columns)
    if faltantes:
        raise KeyError(f"{nome_base}: colunas ausentes: {sorted(faltantes)}")


def taxa_por_100_mil(
    quantidade: pd.Series,
    populacao: pd.Series,
    fator: int = FATOR_TAXA,
) -> pd.Series:
    '''Calcula a taxa de ocorrências padronizada por população.

    Divide a contagem absoluta de eventos pela população correspondente e
    multiplica pelo fator de escala. Valores de população ausentes, nulos ou
    negativos produzem ``NaN`` para evitar distorções.

    Args:
        quantidade: Série numérica com contagens absolutas de ocorrências.
        populacao: Série numérica com a população de referência.
        fator: Constante multiplicadora da taxa (padrão: ``100_000``).

    Returns:
        Série numérica contendo as taxas padronizadas.
    '''
    denominador_valido = populacao.where(populacao > 0)
    return quantidade.div(denominador_valido).mul(fator)


def agregar_contagens_aisp_ano(
    base: pd.DataFrame,
    colunas_contagem: list[str],
) -> pd.DataFrame:
    '''Agrega colunas de contagem por ano e AISP por meio da soma.

    Args:
        base: DataFrame contendo ``ano``, ``id_aisp`` e as colunas de contagem.
        colunas_contagem: Lista de colunas numéricas a serem somadas.

    Returns:
        DataFrame agregado com uma única linha por combinação de ``(ano, id_aisp)``.

    Raises:
        KeyError: Se alguma coluna obrigatória estiver ausente em ``base``.
    '''
    exigir_colunas(base, ["ano", "id_aisp", *colunas_contagem], "base de ocorrências")
    return (
        base.groupby(["ano", "id_aisp"], as_index=False)[colunas_contagem]
        .sum()
    )


def formatar_numero(valor: float, casas: int = 1, sinal: bool = False) -> str:
    '''Formata um valor numérico no padrão brasileiro.

    Utiliza ponto como separador de milhar e vírgula como separador decimal.

    Args:
        valor: Número a ser formatado.
        casas: Quantidade de casas decimais exibidas.
        sinal: Se ``True``, inclui explicitamente o sinal ``+`` em valores positivos.

    Returns:
        String formatada no padrão numérico brasileiro.
    '''
    spec = f"{'+' if sinal else ''},.{casas}f"
    texto = format(valor, spec)
    return texto.replace(",", "_").replace(".", ",").replace("_", ".")


def adicionar_fonte_plotly(
    fig: go.Figure,
    texto: str = FONTE_PADRAO,
    y: float = -0.16,
) -> go.Figure:
    '''Adiciona uma anotação padronizada de fonte no rodapé de uma figura Plotly.

    Args:
        fig: Instância de figura Plotly a ser anotada.
        texto: Texto de atribuição da fonte exibido no rodapé.
        y: Posição vertical relativa em coordenadas de papel (``yref='paper'``).

    Returns:
        A mesma figura Plotly com a anotação de fonte adicionada.
    '''
    fig.add_annotation(
        text=texto,
        xref="paper",
        yref="paper",
        x=0,
        y=y,
        showarrow=False,
        xanchor="left",
        yanchor="top",
        font=dict(size=10, color=PALETA["cinza_medio"]),
    )
    return fig
```

## 2. Carregamento e catálogo das bases

Os arquivos CSV são identificados automaticamente a partir da pasta de dados no repositório do projeto. Cada tabela é preservada separadamente em `dataframes_brutos`.

```python
def carregar_csvs_do_repositorio(
    url_zip: str,
    pasta_dados: str,
    prefixo: str,
    pasta_local: str | None = PASTA_DADOS_LOCAL,
) -> dict[str, pd.DataFrame]:
    '''Lê todos os arquivos CSV a partir da pasta local (se disponível) ou do ZIP remoto.

    Args:
        url_zip: URL direta para o arquivo ``.zip`` do repositório no GitHub.
        pasta_dados: Subcaminho interno no ZIP onde residem os arquivos CSV.
        prefixo: Prefixo comum removido do nome dos arquivos ao gerar as chaves.
        pasta_local: Caminho local opcional utilizado prioritariamente em execuções offline.

    Returns:
        Dicionário mapeando o nome padronizado de cada base ao respectivo DataFrame.

    Raises:
        requests.HTTPError: Se o download do arquivo ZIP falhar.
        ValueError: Se houver nomes de bases duplicados ou nenhum CSV encontrado.
    '''
    bases_carregadas = {}

    if pasta_local and os.path.isdir(pasta_local):
        for nome_arquivo in sorted(os.listdir(pasta_local)):
            if not nome_arquivo.lower().endswith(".csv"):
                continue
            nome_base = nome_arquivo.removesuffix(".csv").removeprefix(prefixo)
            if nome_base in bases_carregadas:
                raise ValueError(f"Nome de base duplicado na pasta local: {nome_base}")
            bases_carregadas[nome_base] = pd.read_csv(
                os.path.join(pasta_local, nome_arquivo), sep=None, engine="python"
            )
        if bases_carregadas:
            return bases_carregadas

    resposta_zip = requests.get(url_zip, timeout=60)
    resposta_zip.raise_for_status()

    with ZipFile(BytesIO(resposta_zip.content)) as arquivo_zip:
        for caminho in arquivo_zip.namelist():
            if pasta_dados not in caminho or not caminho.lower().endswith(".csv"):
                continue

            nome_arquivo = caminho.rsplit("/", 1)[-1]
            nome_base = nome_arquivo.removesuffix(".csv").removeprefix(prefixo)
            if nome_base in bases_carregadas:
                raise ValueError(f"Nome de base duplicado no ZIP: {nome_base}")

            with arquivo_zip.open(caminho) as conteudo_csv:
                bases_carregadas[nome_base] = pd.read_csv(
                    conteudo_csv, sep=None, engine="python"
                )

    if not bases_carregadas:
        raise ValueError(f"Nenhum arquivo CSV encontrado em {pasta_dados!r}.")
    return bases_carregadas


dataframes_brutos = carregar_csvs_do_repositorio(URL_ZIP_DADOS, PASTA_DADOS_ZIP, PREFIXO_CSV)
print(f"{len(dataframes_brutos)} bases importadas:", ", ".join(sorted(dataframes_brutos)))
```

## 3. Limpeza e padronização

Esta rotina produz **cópias** das bases de entrada: `dataframes_brutos` preserva os CSVs importados, enquanto `dataframes` contém as bases tratadas. A etapa padroniza textos e nomes de regiões, preserva valores ausentes de chaves e população e define a coluna temporal `data` nas tabelas mensais.

**Critério de imputação:** ausências em colunas numéricas de contagem iniciadas por `quantidade_` são preenchidas com zero, assumindo ausência de registro de ocorrência naquele recorte.

```python
CORRECOES_REGIAO = {
    "Grande NiterÃƒÂƒÃ‚ÂƒÃƒÂ‚Ã‚Â³i": "Grande Niterói",
    "Grande Niteroi": "Grande Niterói",
}
REGIOES_CANONICAS = {
    "capital": "Capital",
    "baixada fluminense": "Baixada Fluminense",
    "grande niteroi": "Grande Niterói",
    "interior": "Interior",
    "estado": "Estado",
}


def limpar_bases_isp(bases: dict[str, pd.DataFrame]) -> dict[str, pd.DataFrame]:
    '''Padroniza nomes de colunas, textos, tipos temporais e contagens das bases ISP-RJ.

    Cria cópias independentes de cada DataFrame de entrada, remove espaços
    excedentes em colunas textuais, corrige inconsistências de codificação e
    acentuação em ``regiao``, preenche ausências em colunas ``quantidade_*``
    com zero e gera a coluna ``data`` quando ``ano`` e ``mes`` estão presentes.

    Args:
        bases: Dicionário com os DataFrames brutos importados.

    Returns:
        Dicionário com os DataFrames tratados sob as mesmas chaves de entrada.

    Raises:
        ValueError: Se ``ano``, ``mes`` ou ``data`` contiverem valores inválidos.
    '''
    bases_tratadas = {}

    for nome_base, base_original in bases.items():
        base = base_original.copy()
        base.columns = base.columns.str.strip()

        for coluna in base.select_dtypes(include=["object", "string"]).columns:
            base[coluna] = base[coluna].map(
                lambda valor: valor.strip() if isinstance(valor, str) else valor
            )

        if "regiao" in base.columns:
            base["regiao"] = base["regiao"].replace(CORRECOES_REGIAO).map(
                lambda valor: (
                    REGIOES_CANONICAS.get(
                        "".join(
                            ch
                            for ch in unicodedata.normalize("NFKD", valor)
                            if not unicodedata.combining(ch)
                        ).casefold().strip(),
                        valor,
                    )
                    if isinstance(valor, str)
                    else valor
                )
            )

        for coluna in base.select_dtypes(include="number").columns:
            if coluna.startswith("quantidade_"):
                base[coluna] = base[coluna].fillna(0)

        if "ano" in base.columns:
            base["ano"] = pd.to_numeric(base["ano"], errors="raise").astype("int64")
        if "mes" in base.columns:
            base["mes"] = pd.to_numeric(base["mes"], errors="raise").astype("int64")
        if {"ano", "mes"}.issubset(base.columns):
            base["data"] = pd.to_datetime(
                {"year": base["ano"], "month": base["mes"], "day": 1},
                errors="raise"
            )

        bases_tratadas[nome_base] = base

    return bases_tratadas


dataframes = limpar_bases_isp(dataframes_brutos)
print("Bases padronizadas:", len(dataframes))
```

### 5.2. Malhas geográficas (CISP, AISP, RISP)

Centralizamos o carregamento dos shapefiles oficiais (`CISP.zip`, `AISP.zip` e `RISP.zip`) reprojetados para `EPSG:4326` e com chaves padronizadas (`id_cisp`, `id_aisp`, `id_risp`).

```python
def carregar_shapefile_github(
    nome_zip: str,
    chave_padrao: str,
    tolerancia: float = 0.005,
    pasta_local: str | None = PASTA_DADOS_LOCAL,
) -> gpd.GeoDataFrame:
    '''Lê e padroniza um shapefile compactado da pasta local (se disponível) ou do GitHub.

    Padroniza a coluna de chave territorial para ``int64``, reprojeta a malha
    para o sistema de referência ``EPSG:4326`` e simplifica opcionalmente as
    geometrias preservando a topologia.

    Args:
        nome_zip: Nome do arquivo ``.zip`` contendo o shapefile no repositório.
        chave_padrao: Nome padronizado da coluna identificadora (ex.: ``'id_aisp'``).
        tolerancia: Tolerância de simplificação geométrica em graus; se ``None``,
            mantém as geometrias sem simplificação.
        pasta_local: Caminho local opcional utilizado prioritariamente em execuções offline.

    Returns:
        GeoDataFrame padronizado em ``EPSG:4326`` com a coluna ``chave_padrao``.

    Raises:
        requests.HTTPError: Se o download do shapefile falhar.
        KeyError: Se nenhuma coluna candidata à chave territorial for encontrada.
        ValueError: Se o shapefile carregado não possuir CRS definido.
    '''
    caminho_local = os.path.join(pasta_local, nome_zip) if pasta_local else None
    if caminho_local and os.path.isfile(caminho_local):
        gdf = gpd.read_file(f"zip://{caminho_local}")
    else:
        url = f"{URL_BASE_SHAPEFILES}/{nome_zip}"
        resposta = requests.get(url, timeout=60)
        resposta.raise_for_status()

        with tempfile.NamedTemporaryFile(delete=False, suffix=".zip") as tmp:
            tmp.write(resposta.content)
            tmp_path = tmp.name

        try:
            gdf = gpd.read_file(f"zip://{tmp_path}")
        finally:
            os.unlink(tmp_path)

    sufixo = chave_padrao.replace("id_", "")
    candidatos = [chave_padrao, chave_padrao.upper(), sufixo, sufixo.upper()]
    col_encontrada = next((c for c in candidatos if c in gdf.columns), None)
    if col_encontrada is None:
        raise KeyError(
            f"{nome_zip}: nenhuma coluna de chave encontrada entre {candidatos}. Colunas: {list(gdf.columns)}"
        )

    if col_encontrada == chave_padrao:
        gdf[f"{chave_padrao}_original"] = gdf[col_encontrada]
    gdf[chave_padrao] = pd.to_numeric(gdf[col_encontrada], errors="raise").astype("int64")

    if gdf.crs is None:
        raise ValueError(f"{nome_zip}: GeoDataFrame sem CRS definido.")
    if gdf.crs.to_epsg() != 4326:
        gdf = gdf.to_crs(epsg=4326)

    if tolerancia is not None:
        gdf["geometry"] = gdf["geometry"].simplify(tolerance=tolerancia, preserve_topology=True)

    return gdf


gdf_cisp = carregar_shapefile_github("CISP.zip", "id_cisp")
gdf_aisp = carregar_shapefile_github("AISP.zip", "id_aisp")
gdf_risp = carregar_shapefile_github("RISP.zip", "id_risp")

print(f"Shapefiles carregados: CISP={len(gdf_cisp)}, AISP={len(gdf_aisp)}, RISP={len(gdf_risp)}")
```

## 6. Relacionamentos territoriais (CISP, AISP e regiões)

As relações territoriais são construídas **antes** de qualquer cálculo de taxas. `df_rel` registra a correspondência cadastral CISP–AISP–RISP, enquanto para os indicadores históricos extraímos diretamente da base mensal (`df_cisp_mensal`) a AISP observada em cada combinação `(ano, mes, id_cisp)`.

```python
exigir_colunas(df_rel, ["id_cisp", "id_aisp", "regiao"], "df_rel")

grupos = df_rel.groupby("regiao")["id_aisp"].unique()
aisps_por_regiao = {
    regiao: sorted(int(aisp) for aisp in identificadores if pd.notna(aisp))
    for regiao, identificadores in grupos.items()
}

print("Regiões cadastrais disponíveis:", ", ".join(aisps_por_regiao))
```

### 6.1. Dimensão de relacionamento CISP → AISP

Validamos a unicidade da correspondência cadastral e construímos `mapa_cisp_aisp_mensal`, garantindo que nenhuma CISP possua duas AISPs atribuídas no mesmo mês.

```python
df_mapa_cisp_aisp = (
    df_rel[["id_cisp", "id_aisp"]]
    .dropna(subset=["id_cisp", "id_aisp"])
    .drop_duplicates()
    .copy()
)

conflitos_cisp_aisp = (
    df_mapa_cisp_aisp
    .groupby("id_cisp")["id_aisp"]
    .nunique()
    .loc[lambda serie: serie > 1]
)
if not conflitos_cisp_aisp.empty:
    raise ValueError(
        "Uma CISP foi associada a mais de uma AISP. "
        "Verifique a correspondência temporal. "
        f"Exemplos: {conflitos_cisp_aisp.head().to_dict()}"
    )

chave_mensal = ["ano", "mes", "id_cisp"]
mapa_cisp_aisp_mensal = df_cisp_mensal[chave_mensal + ["id_aisp"]].drop_duplicates()
if mapa_cisp_aisp_mensal[chave_mensal + ["id_aisp"]].isna().any().any():
    raise ValueError("Há chaves territoriais mensais ausentes.")
if mapa_cisp_aisp_mensal.duplicated(chave_mensal).any():
    raise ValueError("Há uma CISP com duas AISPs no mesmo mês.")

territorio_aisp1 = (
    df_cisp_mensal.loc[df_cisp_mensal["id_aisp"].eq(1)]
    .groupby("ano")
    .agg(
        cisps=("id_cisp", lambda s: tuple(sorted(s.unique()))),
        meses_observados=("mes", "nunique"),
    )
)
nomes_aisp1 = {
    (6, 7): "Cidade Nova / Santa Teresa (Rio de Janeiro)",
    (72, 73): "São Gonçalo / Neves",
}
territorio_aisp1["nome_aisp"] = territorio_aisp1["cisps"].map(nomes_aisp1).fillna(
    "AISP 1 — composição territorial a verificar"
)
```

### 6.2. Região de policiamento predominante por AISP

Enriquecemos `dim_aisp` com a região de policiamento predominante no período recente (registros a partir de 2018).

```python
regiao_predominante_aisp = (
    df_cisp_mensal.loc[df_cisp_mensal["ano"] >= 2018]
    .groupby("id_aisp")["regiao"]
    .agg(lambda s: s.mode().iloc[0] if not s.mode().empty else "Não informada")
    .rename("regiao_predominante")
    .reset_index()
)

dim_aisp["regiao_predominante"] = dim_aisp["id_aisp"].map(
    regiao_predominante_aisp.set_index("id_aisp")["regiao_predominante"]
)
```

## 7. Padronização da população mensal por CISP

**Granularidade:** `(ano, mes, id_cisp)`. A população mensal é convertida para inteiro (truncando eventuais decimais) antes das agregações anuais, e validamos a ausência de duplicatas ou valores negativos.

```python
exigir_colunas(
    df_populacao_mensal_cisp,
    ["ano", "mes", "id_cisp", "populacao"],
    "df_populacao_mensal_cisp",
)

populacao_texto = (
    df_populacao_mensal_cisp["populacao"]
    .astype("string")
    .str.replace(",", ".", regex=False)
)
populacao_numerica = pd.to_numeric(populacao_texto, errors="raise")
if populacao_numerica.isna().any() or (populacao_numerica < 0).any():
    raise ValueError("Há populações ausentes ou negativas na base mensal por CISP.")

df_populacao_mensal_cisp["populacao"] = populacao_numerica.astype("int64")

duplicatas_mensais = df_populacao_mensal_cisp.duplicated(
    subset=["ano", "mes", "id_cisp"], keep=False,
)
if duplicatas_mensais.any():
    raise ValueError("Há registros duplicados de (ano, mes, id_cisp) na população mensal.")
```

## 8. População anual por CISP

**Granularidade:** `(ano, id_cisp)`. A estimativa anual corresponde à média dos valores mensais disponíveis (sem somar populações entre meses).

```python
df_populacao_anual_cisp = (
    df_populacao_mensal_cisp
    .groupby(["ano", "id_cisp"], as_index=False)["populacao"]
    .mean()
)
df_populacao_anual_cisp.head()
```

## 9. População anual por AISP

Em cada mês, somamos a população das CISPs vinculadas à AISP na base de ocorrências. Um mês só contribui para a média se todas as suas CISPs tiverem população registrada; a estimativa anual é a média desses totais mensais completos. Para comparar taxas anuais, exigimos 12 meses de atividade territorial e população média positiva.

```python
populacao_cisp_com_aisp = mapa_cisp_aisp_mensal.merge(
    df_populacao_mensal_cisp[chave_mensal + ["populacao"]],
    on=chave_mensal, how="outer", validate="one_to_one", indicator="situacao_mapeamento",
)
populacao_cisp_sem_aisp = populacao_cisp_com_aisp.query(
    "situacao_mapeamento == 'right_only'"
).copy()

anos_pop = df_populacao_mensal_cisp["ano"].unique()
pop_mensal_aisp = (
    populacao_cisp_com_aisp.loc[populacao_cisp_com_aisp["ano"].isin(anos_pop)]
    .groupby(["ano", "mes", "id_aisp"], as_index=False)
    .agg(
        populacao=("populacao", "sum"),
        cisps=("id_cisp", "size"),
        cisps_com_pop=("populacao", "count"),
    )
)
pop_mensal_aisp.loc[pop_mensal_aisp["cisps"] != pop_mensal_aisp["cisps_com_pop"], "populacao"] = np.nan

populacao_por_ano_aisp = (
    pop_mensal_aisp.groupby(["ano", "id_aisp"], as_index=False)
    .agg(
        populacao=("populacao", "mean"),
        meses_ativos=("mes", "size"),
        meses_com_populacao=("populacao", "count"),
    )
)
populacao_por_ano_aisp["nome_aisp"] = populacao_por_ano_aisp["id_aisp"].map(nomes_aisp)
linhas_aisp1 = populacao_por_ano_aisp["id_aisp"].eq(1)
populacao_por_ano_aisp.loc[linhas_aisp1, "nome_aisp"] = (
    populacao_por_ano_aisp.loc[linhas_aisp1, "ano"].map(territorio_aisp1["nome_aisp"])
)
populacao_por_ano_aisp.head()
```

## 14. Resumo anual das taxas por AISP

Este bloco sintetiza a **distribuição entre AISPs** (média, mediana, quartis, desvio-padrão populacional com `ddof=0` e coeficiente de variação), que mede a heterogeneidade territorial ao longo do tempo.

```python
def resumir_taxas_por_ano(
    painel: pd.DataFrame,
    coluna_taxa: str = "taxa_letalidade",
) -> pd.DataFrame:
    '''Calcula estatísticas descritivas anuais da distribuição de uma taxa entre AISPs.

    Computa média, mediana, mínimo, máximo, quartis (Q1 e Q3), desvio-padrão
    populacional (``ddof=0``), variância, amplitude interquartil e coeficiente
    de variação percentual por ano.

    Args:
        painel: DataFrame contendo a coluna ``ano`` e a métrica de taxa desejada.
        coluna_taxa: Nome da coluna numérica de taxa a ser sumarizada.

    Returns:
        DataFrame com uma linha por ano e as estatísticas descritivas calculadas.

    Raises:
        KeyError: Se ``ano`` ou ``coluna_taxa`` estiverem ausentes em ``painel``.
    '''
    exigir_colunas(painel, ["ano", coluna_taxa], "painel de taxas")
    resumo = (
        painel.groupby("ano")[coluna_taxa]
        .agg(
            media="mean",
            mediana="median",
            minimo="min",
            maximo="max",
            q1=lambda serie: serie.quantile(0.25),
            q3=lambda serie: serie.quantile(0.75),
            desvio_padrao=lambda serie: serie.std(ddof=0),
        )
        .reset_index()
    )
    resumo["variancia"] = resumo["desvio_padrao"] ** 2
    resumo["amplitude_interquartil"] = resumo["q3"] - resumo["q1"]
    resumo["coef_variacao"] = (
        resumo["desvio_padrao"]
        .div(resumo["media"].where(resumo["media"] != 0))
        .mul(100)
    )
    return resumo


resumo_estatistico = resumir_taxas_por_ano(estatisticas_aisp)
resumo_estatistico.head()
```

## Pergunta 1 — Evolução Histórica das Apreensões de Armas de Fogo e Fuzis no RJ

> **Pergunta de domínio:** *Qual é a evolução histórica do volume absoluto e da taxa (por 100 mil habitantes) de armas de fogo e fuzis apreendidos no estado do Rio de Janeiro nos últimos anos?*

### 1.1 Transformação e Preparação

A partir da base mensal já padronizada `df_armas` (`armas_apreendidas_mensal`), filtramos os anos completos da série histórica (**2007 a 2024**) e agregamos anualmente o total de armas de fogo apreendidas (`quantidade_arma_fogo_total`) e os três principais tipos de armas (`quantidade_fuzil`, `quantidade_pistola` e `quantidade_revolver`).

Para separar o efeito do crescimento demográfico da dinâmica criminal, cruzamos as contagens anuais com `populacao_uf_anual` e calculamos as taxas por 100 mil habitantes utilizando `taxa_por_100_mil`. Diante da ausência de dados populacionais referentes a **2023 e 2024** em `df_letalidade`, aplicamos preenchimento progressivo (*forward fill*, `ffill`), adotando a população de 2022 como referência sob a hipótese de estabilidade populacional no biênio — limitação registrada na conclusão.

Também derivamos:
- **Índice Base 100 (2007 = 100)** das taxas por 100 mil habitantes, comparando diretamente a trajetória relativa do total de armas e dos fuzis em uma mesma escala, sem distorções de eixos duplos;
- **Participação percentual (%)** de revólveres, pistolas e fuzis no total apreendido, para mensurar a substituição tecnológica do arsenal.

```python
colunas_armas_p1 = [
    "quantidade_arma_fogo_total",
    "quantidade_fuzil",
    "quantidade_pistola",
    "quantidade_revolver",
]
exigir_colunas(df_armas, ["ano", *colunas_armas_p1], "df_armas")

armas_anual = (
    df_armas.loc[df_armas["ano"].between(2007, 2024)]
    .groupby("ano", as_index=False)[colunas_armas_p1]
    .sum()
    .sort_values("ano")
    .merge(populacao_uf_anual, on="ano", how="left", validate="one_to_one")
)

anos_pop_ausente = armas_anual.loc[armas_anual["populacao"].isna(), "ano"].tolist()
if anos_pop_ausente:
    print(f"Aviso metodológico: população UF indisponível em df_letalidade para {anos_pop_ausente}; aplicado ffill.")
    armas_anual["populacao"] = armas_anual["populacao"].ffill()

armas_anual["taxa_armas_populacao"] = taxa_por_100_mil(
    armas_anual["quantidade_arma_fogo_total"], armas_anual["populacao"]
)
armas_anual["taxa_fuzil_populacao"] = taxa_por_100_mil(
    armas_anual["quantidade_fuzil"], armas_anual["populacao"]
)

base_armas_2007 = armas_anual.loc[armas_anual["ano"] == 2007, "taxa_armas_populacao"].iloc[0]
base_fuzil_2007 = armas_anual.loc[armas_anual["ano"] == 2007, "taxa_fuzil_populacao"].iloc[0]
armas_anual["indice_taxa_armas"] = (armas_anual["taxa_armas_populacao"] / base_armas_2007) * 100
armas_anual["indice_taxa_fuzil"] = (armas_anual["taxa_fuzil_populacao"] / base_fuzil_2007) * 100

armas_anual["prop_revolver_por_total"] = (
    armas_anual["quantidade_revolver"] / armas_anual["quantidade_arma_fogo_total"]
) * 100
armas_anual["prop_pistola_por_total"] = (
    armas_anual["quantidade_pistola"] / armas_anual["quantidade_arma_fogo_total"]
) * 100
armas_anual["prop_fuzil_por_total"] = (
    armas_anual["quantidade_fuzil"] / armas_anual["quantidade_arma_fogo_total"]
) * 100

resumo_p1 = armas_anual[[
    "ano",
    "quantidade_arma_fogo_total",
    "taxa_armas_populacao",
    "quantidade_fuzil",
    "taxa_fuzil_populacao",
    "prop_revolver_por_total",
    "prop_pistola_por_total",
    "prop_fuzil_por_total",
]].round(2)

display(resumo_p1)
```

### 1.2 Visualização

Para evitar a sobrecarga cognitiva e a distorção visual de eixos Y duplos com ordens de grandeza muito distintas (milhares de armas totais vs. centenas de fuzis), organizamos a evidência em um painel **2×2**: os painéis superiores isolam os volumes absolutos (anotados com as taxas reais nos extremos), o painel inferior esquerdo compara a velocidade relativa das taxas por 100 mil habitantes (Base 100 = 2007) e o painel inferior direito mostra a transição estrutural do arsenal. Linhas verticais tracejadas sinalizam marcos históricos relevantes da segurança pública fluminense — **início das UPPs (2008)**, **crise fiscal do estado (2016)** e **início da pandemia de Covid-19 / ADPF 635 (2020)** — para contextualizar inflexões observadas nas séries.

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 9.8), sharex=True)
(ax_a, ax_b), (ax_c, ax_d) = axes
anos = armas_anual["ano"]

marcos_p1 = [
    (2008, "Início UPPs\n(2008)"),
    (2016, "Crise fiscal\nRJ (2016)"),
    (2020, "Covid-19 /\nADPF 635 (2020)"),
]


def anotar_ponto(
    ax: plt.Axes,
    df: pd.DataFrame,
    col_y: str,
    idx: int,
    texto: str,
    dy: float = 10,
    dx: float = 0,
    cor: str = PALETA["cinza_escuro"],
    ha: str = "center",
) -> None:
    '''Adiciona uma anotação destacada a um ponto específico de um gráfico Matplotlib.

    Args:
        ax: Eixo Matplotlib onde a anotação será inserida.
        df: DataFrame de origem contendo a coluna ``ano`` e ``col_y``.
        col_y: Nome da coluna correspondente à coordenada vertical.
        idx: Rótulo de índice da linha a ser anotada em ``df``.
        texto: Texto exibido na caixa de anotação.
        dy: Deslocamento vertical em pontos a partir do ponto de dados.
        dx: Deslocamento horizontal em pontos a partir do ponto de dados.
        cor: Cor hexadecimal ou nomeada aplicada ao texto da anotação.
        ha: Alinhamento horizontal do texto (``'left'``, ``'center'`` ou ``'right'``).

    Returns:
        None.
    '''
    x_val = df.loc[idx, "ano"]
    y_val = df.loc[idx, col_y]
    ax.annotate(
        texto,
        xy=(x_val, y_val),
        xytext=(dx, dy),
        textcoords="offset points",
        ha=ha,
        fontsize=8.5,
        fontweight="bold",
        color=cor,
        bbox=dict(boxstyle="round,pad=0.2", fc="white", ec="none", alpha=0.85),
        zorder=5,
    )


def adicionar_marcos_historicos(
    ax: plt.Axes,
    marcos: list[tuple[int, str]],
    exibir_rotulos: bool = False,
    y_frac: float = 0.96,
    va: str = "top",
) -> None:
    '''Insere linhas verticais e rótulos opcionais de eventos históricos em um eixo.

    Args:
        ax: Eixo Matplotlib alvo.
        marcos: Lista de tuplas ``(ano, rotulo)`` descrevendo os acontecimentos.
        exibir_rotulos: Se ``True``, desenha o texto de cada marco junto à linha.
        y_frac: Posição vertical relativa do texto em coordenadas do eixo (0 a 1).
        va: Alinhamento vertical do texto (``'top'`` ou ``'bottom'``).

    Returns:
        None.
    '''
    for ano_marco, rotulo_marco in marcos:
        ax.axvline(
            ano_marco,
            color=PALETA["cinza_medio"],
            linestyle=":",
            linewidth=1.1,
            alpha=0.7,
            zorder=1,
        )
        if exibir_rotulos:
            ax.text(
                ano_marco + 0.16,
                y_frac,
                rotulo_marco,
                transform=ax.get_xaxis_transform(),
                fontsize=7.6,
                color=PALETA["cinza_medio"],
                va=va,
                ha="left",
                bbox=dict(boxstyle="round,pad=0.15", fc="white", ec="none", alpha=0.85),
                zorder=4,
            )


ax_a.fill_between(anos, armas_anual["quantidade_arma_fogo_total"], color=PALETA["azul"], alpha=0.10)
ax_a.plot(anos, armas_anual["quantidade_arma_fogo_total"], marker="o", ms=5, lw=2.3, color=PALETA["azul"])
idx_max_a = armas_anual["quantidade_arma_fogo_total"].idxmax()
idx_min_a = armas_anual["quantidade_arma_fogo_total"].idxmin()
anotar_ponto(
    ax_a, armas_anual, "quantidade_arma_fogo_total", idx_max_a,
    (
        f"Pico ({int(armas_anual.loc[idx_max_a, 'ano'])}): "
        f"{formatar_numero(armas_anual.loc[idx_max_a, 'quantidade_arma_fogo_total'], 0)}\n"
        f"({formatar_numero(armas_anual.loc[idx_max_a, 'taxa_armas_populacao'], 1)}/100 mil)"
    ),
    dy=10, dx=8, cor=PALETA["azul"], ha="left",
)
anotar_ponto(
    ax_a, armas_anual, "quantidade_arma_fogo_total", idx_min_a,
    f"Vale ({int(armas_anual.loc[idx_min_a, 'ano'])}): {formatar_numero(armas_anual.loc[idx_min_a, 'quantidade_arma_fogo_total'], 0)}",
    dy=-22, dx=-4, cor=PALETA["cinza_escuro"], ha="right",
)
adicionar_marcos_historicos(ax_a, marcos_p1, exibir_rotulos=True, y_frac=0.05, va="bottom")
ax_a.set_title("(A) Total de Armas de Fogo Apreendidas (Volume Absoluto)", loc="left", pad=10)
ax_a.set_ylabel("Armas apreendidas")
ax_a.yaxis.set_major_formatter(mtick.FuncFormatter(lambda v, _: formatar_numero(v, 0)))
ax_a.set_ylim(0, armas_anual["quantidade_arma_fogo_total"].max() * 1.24)

ax_b.fill_between(anos, armas_anual["quantidade_fuzil"], color=PALETA["vermelho"], alpha=0.10)
ax_b.plot(anos, armas_anual["quantidade_fuzil"], marker="s", ms=5, lw=2.3, color=PALETA["vermelho"])
idx_max_b = armas_anual["quantidade_fuzil"].idxmax()
idx_min_b = armas_anual["quantidade_fuzil"].idxmin()
anotar_ponto(
    ax_b, armas_anual, "quantidade_fuzil", idx_max_b,
    (
        f"Pico ({int(armas_anual.loc[idx_max_b, 'ano'])}): {formatar_numero(armas_anual.loc[idx_max_b, 'quantidade_fuzil'], 0)}\n"
        f"({formatar_numero(armas_anual.loc[idx_max_b, 'taxa_fuzil_populacao'], 2)}/100 mil)"
    ),
    dy=10, dx=-4, cor=PALETA["vermelho"], ha="right",
)
anotar_ponto(
    ax_b, armas_anual, "quantidade_fuzil", idx_min_b,
    f"Vale ({int(armas_anual.loc[idx_min_b, 'ano'])}): {formatar_numero(armas_anual.loc[idx_min_b, 'quantidade_fuzil'], 0)}",
    dy=-20, dx=6, cor=PALETA["cinza_escuro"], ha="left",
)
adicionar_marcos_historicos(ax_b, marcos_p1, exibir_rotulos=True, y_frac=0.05, va="bottom")
ax_b.set_title("(B) Fuzis Apreendidos (Volume Absoluto)", loc="left", pad=10)
ax_b.set_ylabel("Fuzis apreendidos")
ax_b.set_ylim(0, armas_anual["quantidade_fuzil"].max() * 1.28)

ax_c.axhline(100, color=PALETA["cinza_medio"], linestyle=":", lw=1.2, alpha=0.8)
ax_c.plot(anos, armas_anual["indice_taxa_armas"], marker="o", ms=4.5, lw=2.2, color=PALETA["azul"], label="Taxa Total de Armas / 100 mil hab.")
ax_c.plot(anos, armas_anual["indice_taxa_fuzil"], marker="s", ms=4.5, lw=2.4, color=PALETA["vermelho"], label="Taxa de Fuzis / 100 mil hab.")
ult = armas_anual.index[-1]
ax_c.annotate(
    f"  {formatar_numero(armas_anual.loc[ult, 'indice_taxa_armas'], 0)} ({formatar_numero(armas_anual.loc[ult, 'taxa_armas_populacao'], 1)}/100k)",
    xy=(armas_anual.loc[ult, "ano"], armas_anual.loc[ult, "indice_taxa_armas"]),
    va="center", fontsize=8.5, fontweight="bold", color=PALETA["azul"],
)
ax_c.annotate(
    f"  {formatar_numero(armas_anual.loc[ult, 'indice_taxa_fuzil'], 0)} ({formatar_numero(armas_anual.loc[ult, 'taxa_fuzil_populacao'], 2)}/100k)",
    xy=(armas_anual.loc[ult, "ano"], armas_anual.loc[ult, "indice_taxa_fuzil"]),
    va="center", fontsize=8.5, fontweight="bold", color=PALETA["vermelho"],
)
adicionar_marcos_historicos(ax_c, marcos_p1, exibir_rotulos=False)
ax_c.set_title("(C) Evolução Relativa das Taxas por 100 mil hab. (Base 100 = 2007)", loc="left", pad=10)
ax_c.set_ylabel("Índice da taxa (2007 = 100)")
ax_c.set_xlabel("Ano")
ax_c.legend(loc="upper left", fontsize=9.5)

series_comp = [
    ("prop_pistola_por_total", "Pistolas", PALETA["verde"], "o"),
    ("prop_revolver_por_total", "Revólveres", PALETA["laranja"], "^"),
    ("prop_fuzil_por_total", "Fuzis", PALETA["vermelho"], "s"),
]
for col, rotulo, cor, marcador in series_comp:
    ax_d.plot(anos, armas_anual[col], marker=marcador, ms=4.5, lw=2.2, color=cor, label=rotulo)
    ax_d.annotate(
        f"  {rotulo}: {formatar_numero(armas_anual.loc[ult, col], 1)}%",
        xy=(armas_anual.loc[ult, "ano"], armas_anual.loc[ult, col]),
        va="center", fontsize=8.5, fontweight="bold", color=cor,
    )

adicionar_marcos_historicos(ax_d, marcos_p1, exibir_rotulos=False)
ax_d.set_title("(D) Transição Qualitativa do Arsenal Apreendido (% do Total)", loc="left", pad=10)
ax_d.set_ylabel("Participação no total (%)")
ax_d.set_xlabel("Ano")
ax_d.set_ylim(0, 62)

for ax in axes.flat:
    ax.set_xticks(range(2007, 2025, 2))
    ax.set_xlim(2006.5, 2026.8)
    ax.grid(axis="x", visible=False)

fig.suptitle(
    "Apreensão total de armas recua no longo prazo, mas fuzis e pistolas ganham peso no arsenal do RJ (2007–2024)",
    fontsize=13.5, fontweight="bold", y=0.99, color=PALETA["cinza_escuro"],
)
fig.text(
    0.01, 0.005,
    f"{FONTE_PADRAO} Linhas verticais pontilhadas indicam marcos contextuais (2008, 2016 e 2020).",
    fontsize=9, color=PALETA["cinza_medio"],
)
plt.tight_layout(rect=[0, 0.025, 1, 0.95])
plt.show()
```

> **Pergunta de domínio:** *Quais municípios ou Áreas Integradas (AISP) concentram as maiores taxas proporcionais de letalidade violenta e feminicídio?*

### 2.1 Transformação e Preparação

Optamos pelo recorte por **AISP** por dois motivos: a base oficial de feminicídio (`df_fem_cisp`) é disponibilizada em granularidade CISP/AISP e esse nível territorial permite desagregar a Capital (que, no recorte municipal, apareceria como um único bloco heterogêneo). Utilizamos `estatisticas_aisp` para a letalidade violenta e construímos um painel paralelo de feminicídio na mesma granularidade `(ano, id_aisp)` com `agregar_contagens_aisp_ano`.

Três cuidados metodológicos são essenciais nesta etapa:

1. **Cobertura temporal do feminicídio:** os registros oficiais de `df_fem_cisp` iniciam-se em **novembro de 2016** (apenas 2 meses de coleta, totalizando 14 casos no estado, contra 68 em 2017). Para evitar distorção na taxa anual e na comparação temporal, restringimos o cálculo de taxas de feminicídio aos **anos completos a partir de 2017** que possuem denominador populacional em `populacao_por_ano_aisp`, exigindo 12 meses de atividade territorial (`meses_ativos == 12`).

2. **Adequação do denominador populacional:** diferentemente da letalidade violenta, cuja taxa é calculada em relação à população total, a taxa de feminicídio deve considerar a população feminina exposta. Diante da ausência de população por sexo na base utilizada, adotamos a aproximação de que a população feminina corresponde a **50% da população total de cada AISP**:

   $$
   T_{\text{feminicídio}}
   = \frac{F}{P/2} \times 100.000
   = \frac{F}{P} \times 200.000
   $$

   Utilizamos `fator=200_000` em `taxa_por_100_mil` para obter taxas por **100 mil mulheres estimadas**, mantendo `100_000` para a letalidade violenta por **100 mil habitantes**.

3. **Estabilidade estatística em eventos raros:** como o feminicídio apresenta contagens anuais reduzidas em cada AISP, calculamos a **taxa média anual ponderada por pessoas-ano** no período comum (`2017–2022`), reduzindo a variabilidade de anos isolados:

   $$
   T_{\text{letalidade}}
   = \frac{\sum \text{ocorrências}}{\sum \text{população}} \times 100.000
   \qquad \text{e} \qquad
   T_{\text{feminicídio}}
   = \frac{\sum \text{feminicídios}}{\sum \text{população}} \times 200.000
   $$

```python
exigir_colunas(
    df_fem_cisp,
    ["ano", "id_aisp", "quantidade_morte_feminicidio", "quantidade_tentativa_feminicidio"],
    "df_fem_cisp",
)

fem_aisp_ano = agregar_contagens_aisp_ano(
    df_fem_cisp,
    ["quantidade_morte_feminicidio", "quantidade_tentativa_feminicidio"],
)

ultimo_ano_pop = int(populacao_por_ano_aisp["ano"].max())
ano_inicio_fem_completo = 2017

painel_fem_aisp = (
    populacao_por_ano_aisp.loc[
        populacao_por_ano_aisp["ano"].between(ano_inicio_fem_completo, ultimo_ano_pop)
    ]
    .merge(fem_aisp_ano, on=["ano", "id_aisp"], how="left", validate="one_to_one")
    .merge(
        dim_aisp[["id_aisp", "regiao_predominante", "rotulo_aisp"]],
        on="id_aisp",
        how="left",
        validate="many_to_one",
    )
)

painel_fem_aisp["quantidade_morte_feminicidio"] = (
    painel_fem_aisp["quantidade_morte_feminicidio"].fillna(0)
)

painel_fem_aisp["taxa_feminicidio"] = taxa_por_100_mil(
    painel_fem_aisp["quantidade_morte_feminicidio"],
    painel_fem_aisp["populacao"].where(painel_fem_aisp["meses_ativos"].eq(12)),
    fator=200_000,
)

painel_let_aisp = (
    estatisticas_aisp.dropna(subset=["taxa_letalidade"])
    .merge(
        dim_aisp[["id_aisp", "regiao_predominante", "rotulo_aisp"]],
        on="id_aisp",
        how="left",
        validate="many_to_one",
    )
    .copy()
)

for painel in (painel_let_aisp, painel_fem_aisp):
    linhas = painel["id_aisp"].eq(1)
    nomes = painel.loc[linhas, "ano"].map(territorio_aisp1["nome_aisp"])
    painel.loc[linhas, "nome_aisp"] = nomes
    painel.loc[linhas, "rotulo_aisp"] = "AISP 01 — " + nomes
    painel.loc[linhas, "regiao_predominante"] = painel.loc[linhas, "ano"].map(
        df_cisp_mensal.loc[df_cisp_mensal["id_aisp"].eq(1)]
        .groupby("ano")["regiao"].agg(lambda s: s.mode().iloc[0])
    )

let_periodo_comum = (
    painel_let_aisp.loc[
        painel_let_aisp["ano"].between(ano_inicio_fem_completo, ultimo_ano_pop)
    ]
    .groupby(
        ["id_aisp", "nome_aisp", "rotulo_aisp", "regiao_predominante"],
        as_index=False,
    )
    .agg(
        total_letalidade=("quantidade_letalidade_violenta", "sum"),
        pessoas_ano=("populacao", "sum"),
    )
)

let_periodo_comum["taxa_media_letalidade"] = taxa_por_100_mil(
    let_periodo_comum["total_letalidade"],
    let_periodo_comum["pessoas_ano"],
)

fem_periodo_comum = (
    painel_fem_aisp.loc[painel_fem_aisp["meses_ativos"].eq(12)]
    .groupby("id_aisp", as_index=False)
    .agg(total_feminicidio=("quantidade_morte_feminicidio", "sum"))
)

comparativo_aisp_p2 = let_periodo_comum.merge(
    fem_periodo_comum,
    on="id_aisp",
    how="left",
    validate="one_to_one",
)

comparativo_aisp_p2["total_feminicidio"] = (
    comparativo_aisp_p2["total_feminicidio"].fillna(0)
)

comparativo_aisp_p2["taxa_media_feminicidio"] = taxa_por_100_mil(
    comparativo_aisp_p2["total_feminicidio"],
    comparativo_aisp_p2["pessoas_ano"],
    fator=200_000,
)

r_espacial_let_fem = comparativo_aisp_p2["taxa_media_letalidade"].corr(
    comparativo_aisp_p2["taxa_media_feminicidio"]
)

fem_anual_uf = (
    df_fem_cisp.groupby("ano", as_index=False)["quantidade_morte_feminicidio"]
    .sum()
)

let_anual_uf = (
    mortes_aisp_ano.groupby("ano", as_index=False)["quantidade_letalidade_violenta"]
    .sum()
)

serie_let_fem_uf = (
    fem_anual_uf.merge(let_anual_uf, on="ano", validate="one_to_one")
    .query("2017 <= ano <= 2025")
    .sort_values("ano")
)

r_temporal_let_fem_2017_2025 = serie_let_fem_uf["quantidade_morte_feminicidio"].corr(
    serie_let_fem_uf["quantidade_letalidade_violenta"]
)

print(f"Período comum de taxas por AISP (com anos completos de feminicídio): {ano_inicio_fem_completo}–{ultimo_ano_pop}")
print(f"Correlação espacial entre taxa média de letalidade e feminicídio por AISP (r): {r_espacial_let_fem:.4f}")
print(f"Correlação temporal estadual Letalidade vs. Feminicídio (2017–2025, r): {r_temporal_let_fem_2017_2025:.4f}")

print("\nTop 10 AISPs — Taxa média anual de Letalidade Violenta (por 100 mil habitantes):")
display(
    comparativo_aisp_p2.nlargest(10, "taxa_media_letalidade")[
        [
            "rotulo_aisp",
            "regiao_predominante",
            "taxa_media_letalidade",
            "total_letalidade",
            "taxa_media_feminicidio",
        ]
    ].round(2)
)

print("\nTop 10 AISPs — Taxa média anual de Feminicídio (por 100 mil mulheres estimadas):")
display(
    comparativo_aisp_p2.nlargest(10, "taxa_media_feminicidio")[
        [
            "rotulo_aisp",
            "regiao_predominante",
            "taxa_media_feminicidio",
            "total_feminicidio",
            "taxa_media_letalidade",
        ]
    ].round(2)
)
```

## Pergunta 3 — Sazonalidade Mensal da Letalidade Violenta por Região de Policiamento

> **Pergunta de domínio:** *Como se comporta a sazonalidade da letalidade violenta ao longo dos meses do ano nas diferentes regiões de policiamento do estado?*

### 3.1 Transformação e Preparação

A partir de `df_cisp_mensal`, filtramos os anos completos da série histórica (**2003 a 2024**) e agregamos `quantidade_letalidade_violenta` por `regiao` (`Capital`, `Baixada Fluminense`, `Grande Niterói` e `Interior`) e `mes`, além de construir a linha de referência estadual (`Estado (RJ)`).

**Ajuste de calendário e Índice Sazonal (Base 100):** meses possuem durações diferentes (fevereiro tem em média 28,25 dias, contra 31 dias em janeiro e março — diferença próxima de 10%). Para evitar que fevereiro pareça artificialmente menos violento apenas por ser mais curto, convertemos a soma mensal em **média diária de vítimas** dividindo pelo número médio de dias de cada mês e calculamos o **Índice Sazonal (Base 100)** em relação à média diária anual de cada região:
- Valores **acima de 100** indicam meses com intensidade diária acima da média daquela região;
- Valores **abaixo de 100** indicam meses relativamente mais brandos.

```python
exigir_colunas(df_cisp_mensal, ["ano", "mes", "regiao", "quantidade_letalidade_violenta"], "df_cisp_mensal")

regioes_ordem_p3 = ["Capital", "Baixada Fluminense", "Grande Niterói", "Interior"]
meses_rotulos = ["Jan", "Fev", "Mar", "Abr", "Mai", "Jun", "Jul", "Ago", "Set", "Out", "Nov", "Dez"]
base_p3 = df_cisp_mensal.loc[
    df_cisp_mensal["ano"].between(2003, 2024)
    & df_cisp_mensal["regiao"].isin(regioes_ordem_p3)
].copy()

anos_meses_p3 = base_p3[["ano", "mes"]].drop_duplicates().copy()
anos_meses_p3["dias_mes"] = pd.to_datetime(
    {"year": anos_meses_p3["ano"], "month": anos_meses_p3["mes"], "day": 1}
).dt.days_in_month
dias_por_mes = (
    anos_meses_p3.groupby("mes")["dias_mes"]
    .mean()
    .reindex(range(1, 13))
)

sazonal_reg = (
    base_p3.groupby(["regiao", "mes"])["quantidade_letalidade_violenta"]
    .sum()
    .unstack("mes")
    .reindex(index=regioes_ordem_p3, columns=range(1, 13))
)
sazonal_estado = (
    base_p3.groupby("mes")["quantidade_letalidade_violenta"]
    .sum()
    .reindex(range(1, 13))
    .rename("Estado (RJ)")
    .to_frame()
    .T
)
sazonal_bruto = pd.concat([sazonal_reg, sazonal_estado], axis=0)

intensidade_diaria = sazonal_bruto.div(dias_por_mes, axis=1)
indice_sazonal = intensidade_diaria.div(intensidade_diaria.mean(axis=1), axis=0).mul(100)
indice_sazonal.columns = meses_rotulos

resumo_p3 = pd.DataFrame({
    "mes_pico": indice_sazonal.idxmax(axis=1),
    "indice_pico": indice_sazonal.max(axis=1).round(2),
    "mes_vale": indice_sazonal.idxmin(axis=1),
    "indice_vale": indice_sazonal.min(axis=1).round(2),
    "amplitude_pts": (indice_sazonal.max(axis=1) - indice_sazonal.min(axis=1)).round(2),
})

print("Índice Sazonal Base 100 (ajustado pela duração dos meses, 2003–2024):")
display(indice_sazonal.round(1))
print("\nResumo de picos, vales e amplitude sazonal:")
display(resumo_p3)
```

## Pergunta 4 — Evolução Temporal da Taxa de Letalidade: Capital, Entorno Metropolitano e Interior

> **Pergunta de domínio:** *Como se compara a evolução temporal da taxa de letalidade entre a capital (Rio de Janeiro), a Região Metropolitana e as regiões do interior?*

### 4.1 Transformação e Preparação

Utilizamos a base oficial `df_letalidade` (`taxa_letalidade`), filtrada para `delito == "Letalidade Violenta"`, empregando diretamente a série padronizada `taxa_cem_mil_habitantes` para as quatro regiões oficiais de policiamento (`Capital`, `Baixada Fluminense`, `Grande Niterói` — sendo estas duas últimas o entorno metropolitano — e `Interior`) além do agregado `Estado`.

**Nota de cobertura temporal:** na base oficial do ISP-RJ, `Capital` e `Estado` iniciam-se em **1991**, enquanto a desagregação completa das quatro regiões existe desde **2000**. Preservamos o horizonte longo (`1991–2022`) no gráfico de linhas interativo e adotamos a janela **`2003–2022`** (alinhada ao recorte temporal das séries por CISP/AISP utilizadas nas demais seções) para as comparações diretas de variação percentual e convergência (razão Máx/Mín entre regiões).

```python
exigir_colunas(
    df_letalidade,
    ["ano", "regiao", "delito", "contagem_delito", "populacao", "taxa_cem_mil_habitantes"],
    "df_letalidade",
)

quatro_reg = ["Capital", "Baixada Fluminense", "Grande Niterói", "Interior"]


def normalizar_regiao_p4(valor: object) -> str:
    '''Normaliza o nome de uma região removendo acentos e padronizando caixa.

    Args:
        valor: Texto ou valor representando o nome da região na base de dados.

    Returns:
        String normalizada em minúsculas, sem acentuação e sem espaços extras.
    '''
    texto = unicodedata.normalize("NFKD", str(valor))
    return "".join(ch for ch in texto if not unicodedata.combining(ch)).casefold().strip()


canonico_p4 = {normalizar_regiao_p4(r): r for r in [*quatro_reg, "Estado"]}
df_letalidade_p4 = df_letalidade.copy()
df_letalidade_p4["regiao_p4"] = df_letalidade_p4["regiao"].map(
    lambda valor: canonico_p4.get(normalizar_regiao_p4(valor))
)
cols_resumo_p4 = [*quatro_reg, "Estado"]

df_let_p4 = (
    df_letalidade_p4.loc[
        (df_letalidade_p4["delito"] == "Letalidade Violenta")
        & (df_letalidade_p4["regiao_p4"].isin(cols_resumo_p4))
    ]
    .assign(regiao=lambda base: base["regiao_p4"])
    .sort_values(["regiao", "ano"])
    .copy()
)

piv_p4 = (
    df_let_p4.pivot(index="ano", columns="regiao", values="taxa_cem_mil_habitantes")
    .sort_index()
)

anos_comuns_p4 = piv_p4[quatro_reg].dropna().index
ano_ini_comum_p4, ano_fim_comum_p4 = int(anos_comuns_p4.min()), int(anos_comuns_p4.max())

anos_marco = [1991, 1995, 2003, 2008, 2012, 2017, 2020, 2022]
resumo_p4_marcos = piv_p4.loc[piv_p4.index.isin(anos_marco), cols_resumo_p4].round(2)

resumo_p4_extremos = pd.DataFrame({
    "taxa_2003": piv_p4.loc[2003, cols_resumo_p4],
    "ano_pico": piv_p4[cols_resumo_p4].idxmax(),
    "taxa_pico": piv_p4[cols_resumo_p4].max(),
    "ano_vale": piv_p4[cols_resumo_p4].idxmin(),
    "taxa_vale": piv_p4[cols_resumo_p4].min(),
    "taxa_2022": piv_p4.loc[2022, cols_resumo_p4],
    "var_2003_2022_pct": ((piv_p4.loc[2022, cols_resumo_p4] / piv_p4.loc[2003, cols_resumo_p4]) - 1) * 100,
}).round(2)

piv_comum = piv_p4.loc[anos_comuns_p4, quatro_reg]
media_regional_p4 = piv_comum.mean(axis=1)
desvio_regional_p4 = piv_comum.std(axis=1, ddof=0)
cv_regioes_p4 = (
    desvio_regional_p4.div(media_regional_p4.where(media_regional_p4.ne(0)))
    .mul(100)
    .rename("coef_variacao")
    .reset_index()
)
razao_max_min_p4 = piv_comum.max(axis=1).div(piv_comum.min(axis=1))

df_let_p4["contagem_p4"] = (
    df_let_p4["taxa_cem_mil_habitantes"].mul(df_let_p4["populacao"])
    .div(FATOR_TAXA)
    .round()
)

contagens_regioes_p4 = (
    df_let_p4.loc[df_let_p4["regiao"].isin(quatro_reg)]
    .pivot(index="ano", columns="regiao", values="contagem_p4")
    .reindex(index=anos_comuns_p4, columns=quatro_reg)
)
if contagens_regioes_p4.isna().any().any():
    raise ValueError("Há contagens ausentes para uma das quatro regiões.")
totais_regionais_p4 = contagens_regioes_p4.sum(axis=1)
if totais_regionais_p4.le(0).any():
    raise ValueError("Há total regional nulo ou negativo.")
proporcoes_p4 = contagens_regioes_p4.div(totais_regionais_p4, axis=0).mul(100)
if not np.allclose(proporcoes_p4.sum(axis=1), 100):
    raise ValueError("As participações regionais não somam 100% em todos os anos.")
proporcao_regional_p4 = (
    proporcoes_p4.rename_axis("ano")
    .reset_index()
    .melt(id_vars="ano", var_name="regiao", value_name="proporcao_percentual")
)

print(f"Janela comum das 4 regiões: {ano_ini_comum_p4}–{ano_fim_comum_p4}")
print("\nMarcos históricos da Taxa de Letalidade Violenta por 100 mil hab. (1991–2022):")
display(resumo_p4_marcos)
print("\nExtremos na série completa (1991–2022 / 2000–2022) e variação acumulada (2003–2022):")
display(resumo_p4_extremos)
print(
    f"\nConvergência regional (2003 → 2022): "
    f"Razão Máx/Mín = {razao_max_min_p4.loc[2003]:.2f}x → {razao_max_min_p4.loc[2022]:.2f}x | "
    f"CV regional = {cv_regioes_p4.loc[cv_regioes_p4['ano'] == 2003, 'coef_variacao'].iloc[0]:.1f}% → "
    f"{cv_regioes_p4.loc[cv_regioes_p4['ano'] == 2022, 'coef_variacao'].iloc[0]:.1f}%"
)
```

## Pergunta 5 — Clusters Territoriais: Intervenção Policial e Apreensão de Armamento Pesado

> **Pergunta de domínio:** *O mapeamento territorial revela agrupamentos (clusters) onde altas taxas de mortes por intervenção policial coexistem de forma consistente com volumes elevados de apreensão de armamento pesado?*

### 5.1 Transformação e Preparação

Para testar a coexistência territorial entre letalidade policial e armamento de guerra sem distorções pontuais de um único ano, consolidamos o período recente com cobertura populacional completa (**2017–2022**, alinhado à Pergunta 2) por `id_aisp`, reutilizando `agregar_contagens_aisp_ano`, `taxa_por_100_mil`, `populacao_por_ano_aisp` e `dim_aisp`:
- **Eixo Y (Letalidade Policial):** taxa média anual de mortes por intervenção de agente do Estado por 100 mil habitantes (`quantidade_homicidio_intervencao_policial` dividida por pessoas-ano);
- **Eixo X (Armamento Pesado):** volume acumulado de fuzis apreendidos (`quantidade_fuzil`), acompanhado da taxa proporcional de fuzis (`taxa_fuzil_populacao`) como verificação de robustez.

Classificamos as **39 AISPs com cobertura populacional completa no período** (entre os 41 polígonos da malha; AISPs 1 e 43 sem série populacional em 2017–2022) em **quatro quadrantes (clusters)** definidos pelas **medianas** de cada variável (critério robusto a valores extremos), destacando o **Cluster 1 (Foco Crítico: Alta MDIP + Alto Volume de Fuzis)**.

```python
exigir_colunas(df_cisp_mensal, ["ano", "id_aisp", "quantidade_homicidio_intervencao_policial"], "df_cisp_mensal")
exigir_colunas(df_armas, ["ano", "id_aisp", "quantidade_fuzil"], "df_armas")

ano_ini_p5, ano_fim_p5 = 2017, int(populacao_por_ano_aisp["ano"].max())

mdip_aisp_ano = agregar_contagens_aisp_ano(
    df_cisp_mensal.loc[df_cisp_mensal["ano"].between(ano_ini_p5, ano_fim_p5)],
    ["quantidade_homicidio_intervencao_policial"],
)
fuzil_aisp_ano = agregar_contagens_aisp_ano(
    df_armas.loc[df_armas["ano"].between(ano_ini_p5, ano_fim_p5)],
    ["quantidade_fuzil"],
)

pop_p5 = (
    populacao_por_ano_aisp.loc[
        populacao_por_ano_aisp["ano"].between(ano_ini_p5, ano_fim_p5)
        & populacao_por_ano_aisp["meses_ativos"].eq(12)
    ]
    .groupby("id_aisp", as_index=False)
    .agg(pessoas_ano=("populacao", "sum"), pop_media=("populacao", "mean"))
)

painel_p5 = (
    pop_p5
    .merge(
        mdip_aisp_ano.groupby("id_aisp", as_index=False)["quantidade_homicidio_intervencao_policial"].sum(),
        on="id_aisp", how="left", validate="one_to_one",
    )
    .merge(
        fuzil_aisp_ano.groupby("id_aisp", as_index=False)["quantidade_fuzil"].sum(),
        on="id_aisp", how="left", validate="one_to_one",
    )
    .merge(
        dim_aisp[["id_aisp", "nome_aisp", "rotulo_aisp", "regiao_predominante"]],
        on="id_aisp", how="left", validate="one_to_one",
    )
)

painel_p5["quantidade_homicidio_intervencao_policial"] = painel_p5["quantidade_homicidio_intervencao_policial"].fillna(0)
painel_p5["quantidade_fuzil"] = painel_p5["quantidade_fuzil"].fillna(0)

painel_p5["taxa_mdip_100k"] = taxa_por_100_mil(
    painel_p5["quantidade_homicidio_intervencao_policial"], painel_p5["pessoas_ano"]
)
painel_p5["taxa_fuzil_populacao"] = taxa_por_100_mil(
    painel_p5["quantidade_fuzil"], painel_p5["pessoas_ano"]
)

mediana_fuzil = float(painel_p5["quantidade_fuzil"].median())
mediana_mdip = float(painel_p5["taxa_mdip_100k"].median())


def classificar_cluster(row: pd.Series) -> str:
    '''Classifica uma AISP em um dos quatro quadrantes de confronto armado.

    Compara o volume acumulado de fuzis apreendidos e a taxa média anual de
    mortes por intervenção policial (MDIP) com suas respectivas medianas.

    Args:
        row: Linha do painel territorial contendo ``quantidade_fuzil`` e
            ``taxa_mdip_100k``.

    Returns:
        Rótulo textual do cluster territorial correspondente (Cluster 1 a 4).
    '''
    alto_fuzil = row["quantidade_fuzil"] >= mediana_fuzil
    alta_mdip = row["taxa_mdip_100k"] >= mediana_mdip
    if alto_fuzil and alta_mdip:
        return "Cluster 1: Alta MDIP + Alto Fuzil"
    if (not alto_fuzil) and alta_mdip:
        return "Cluster 2: Alta MDIP + Baixo Fuzil"
    if alto_fuzil and (not alta_mdip):
        return "Cluster 3: Baixa MDIP + Alto Fuzil"
    return "Cluster 4: Baixa MDIP + Baixo Fuzil"


painel_p5["cluster"] = painel_p5.apply(classificar_cluster, axis=1)

r_pearson_vol = painel_p5["quantidade_fuzil"].corr(painel_p5["taxa_mdip_100k"], method="pearson")
r_spearman_vol = painel_p5["quantidade_fuzil"].corr(painel_p5["taxa_mdip_100k"], method="spearman")
r_spearman_taxa = painel_p5["taxa_fuzil_populacao"].corr(painel_p5["taxa_mdip_100k"], method="spearman")

print(f"Período consolidado P5: {ano_ini_p5}–{ano_fim_p5} ({len(painel_p5)} AISPs)")
print(f"Medianas de corte: Fuzis acumulados = {mediana_fuzil:.1f} | Taxa MDIP = {mediana_mdip:.2f} / 100 mil hab.")
print(f"Correlação Volume de Fuzis vs. Taxa MDIP: Pearson r = {r_pearson_vol:.4f} | Spearman rho = {r_spearman_vol:.4f}")
print(f"Correlação Taxa de Fuzis vs. Taxa MDIP (robustez): Spearman rho = {r_spearman_taxa:.4f}")
print("\nDistribuição das AISPs por cluster:")
display(painel_p5["cluster"].value_counts().rename_axis("cluster").reset_index(name="qtd_aisps"))

print("\nAISPs no Cluster 1 (Foco Crítico — ordenadas por Taxa MDIP):")
display(
    painel_p5.loc[painel_p5["cluster"] == "Cluster 1: Alta MDIP + Alto Fuzil"]
    .sort_values("taxa_mdip_100k", ascending=False)[
        ["rotulo_aisp", "regiao_predominante", "taxa_mdip_100k", "quantidade_homicidio_intervencao_policial", "quantidade_fuzil", "taxa_fuzil_populacao"]
    ]
    .round(2)
)
```

## Pergunta 6 — Peso das Intervenções Policiais na Letalidade dos Municípios Mais Populosos

> **Pergunta de domínio:** *A variação na proporção de mortes por intervenção policial afeta a tendência geral de letalidade violenta a longo prazo nos municípios mais populosos?*

### 6.1 Transformação e Preparação

Integramos `df_mun_mensal` (`quantidade_letalidade_violenta`, `quantidade_homicidio_intervencao_policial`, `quantidade_registro_ocorrencia`) com `df_taxa_mun_mensal` (`taxa_letalidade_violenta`, `taxa_registro_ocorrencia`) no período comum com anos completos nas duas bases (**2014 a 2024**):
1. Estimamos a população média municipal pela identidade oficial $\text{pop} = \frac{\text{quantidade\_registro\_ocorrencia}}{\text{taxa\_registro\_ocorrencia}} \times 100.000$ — procedimento que permite estender a seleção dos **6 municípios mais populosos** até 2024, superando o corte em 2022 das bases populacionais isoladas;
2. Agregamos anualmente, por município, a **taxa anual de letalidade violenta por 100 mil habitantes** (soma das 12 taxas mensais oficiais em `df_taxa_mun_mensal`, aproximação consistente sob baixa variação populacional intra-anual) e a **participação percentual (%) das mortes por intervenção policial** na letalidade violenta (`prop_mdip_pct = MDIP / Letalidade Violenta × 100`);
3. Calculamos a correlação temporal de Pearson ($r$) entre `prop_mdip_pct` e `taxa_letalidade_anual` em cada município — tanto em nível quanto em variação anual ($\Delta$) — para avaliar se picos de participação policial acompanham alta ou queda da letalidade geral.

```python
exigir_colunas(
    df_mun_mensal,
    ["ano", "mes", "id_municipio", "quantidade_letalidade_violenta", "quantidade_homicidio_intervencao_policial", "quantidade_registro_ocorrencia"],
    "df_mun_mensal",
)
exigir_colunas(
    df_taxa_mun_mensal,
    ["ano", "mes", "id_municipio", "taxa_letalidade_violenta", "taxa_registro_ocorrencia"],
    "df_taxa_mun_mensal",
)

NOMES_MUNICIPIOS_RJ = {
    3304557: "Rio de Janeiro",
    3304904: "São Gonçalo",
    3301702: "Duque de Caxias",
    3303500: "Nova Iguaçu",
    3303302: "Niterói",
    3300456: "Belford Roxo",
    3301009: "Campos dos Goytacazes",
    3305109: "São João de Meriti",
    3303906: "Petrópolis",
    3306305: "Volta Redonda",
}

base_mun_p6 = df_mun_mensal[
    ["ano", "mes", "id_municipio", "quantidade_letalidade_violenta", "quantidade_homicidio_intervencao_policial", "quantidade_registro_ocorrencia"]
].merge(
    df_taxa_mun_mensal[["ano", "mes", "id_municipio", "taxa_letalidade_violenta", "taxa_registro_ocorrencia"]],
    on=["ano", "mes", "id_municipio"],
    how="inner",
    validate="one_to_one",
)

meses_por_ano_mun = base_mun_p6.groupby("ano")["mes"].nunique()
anos_completos_p6 = meses_por_ano_mun.loc[meses_por_ano_mun == 12].index
base_mun_p6 = base_mun_p6.loc[base_mun_p6["ano"].isin(anos_completos_p6)].copy()

taxa_ro_valida = base_mun_p6["taxa_registro_ocorrencia"].where(base_mun_p6["taxa_registro_ocorrencia"] > 0)
base_mun_p6["pop_estimada"] = (base_mun_p6["quantidade_registro_ocorrencia"] / taxa_ro_valida) * FATOR_TAXA

ano_ref_pop = int(base_mun_p6["ano"].max())
top6_ids = (
    base_mun_p6.loc[base_mun_p6["ano"] >= ano_ref_pop - 2]
    .groupby("id_municipio")["pop_estimada"]
    .mean()
    .nlargest(6)
    .index.tolist()
)

painel_p6 = (
    base_mun_p6.loc[base_mun_p6["id_municipio"].isin(top6_ids)]
    .groupby(["ano", "id_municipio"], as_index=False)
    .agg(
        letalidade=("quantidade_letalidade_violenta", "sum"),
        mdip=("quantidade_homicidio_intervencao_policial", "sum"),
        taxa_letalidade_anual=("taxa_letalidade_violenta", "sum"),
        pop_media=("pop_estimada", "mean"),
    )
)
painel_p6["nome_municipio"] = (
    painel_p6["id_municipio"].map(NOMES_MUNICIPIOS_RJ).fillna(painel_p6["id_municipio"].astype(str))
)
painel_p6["prop_mdip_pct"] = np.where(
    painel_p6["letalidade"] > 0,
    (painel_p6["mdip"] / painel_p6["letalidade"]) * 100,
    np.nan,
)
painel_p6 = painel_p6.sort_values(["id_municipio", "ano"]).reset_index(drop=True)

ordem_mun_p6 = [NOMES_MUNICIPIOS_RJ.get(i, str(i)) for i in top6_ids]
linhas_resumo_p6 = []
for mun in ordem_mun_p6:
    sub = painel_p6.loc[painel_p6["nome_municipio"] == mun].sort_values("ano")
    r_nivel = sub["prop_mdip_pct"].corr(sub["taxa_letalidade_anual"])
    r_diff = sub["prop_mdip_pct"].diff().corr(sub["taxa_letalidade_anual"].diff())
    linhas_resumo_p6.append({
        "municipio": mun,
        "ano_ini": int(sub["ano"].min()),
        "ano_fim": int(sub["ano"].max()),
        "taxa_ini": sub["taxa_letalidade_anual"].iloc[0],
        "taxa_fim": sub["taxa_letalidade_anual"].iloc[-1],
        "prop_mdip_ini_pct": sub["prop_mdip_pct"].iloc[0],
        "prop_mdip_fim_pct": sub["prop_mdip_pct"].iloc[-1],
        "prop_mdip_max_pct": sub["prop_mdip_pct"].max(),
        "ano_pico_mdip": int(sub.loc[sub["prop_mdip_pct"].idxmax(), "ano"]),
        "r_nivel": r_nivel,
        "r_variacao_anual": r_diff,
    })

resumo_p6 = pd.DataFrame(linhas_resumo_p6).round(2)
print(f"Período completo de municípios analisado: {int(painel_p6['ano'].min())}–{int(painel_p6['ano'].max())}")
display(resumo_p6)
```

Já em **variações anuais ($\Delta$)**, a correlação é positiva em **todos os seis municípios** (de **$+0{,}22$** em São Gonçalo e **$+0{,}27$** na Capital até **$+0{,}61$** em Niterói): nos anos em que a participação policial acelerou, a taxa geral tendeu a repicar no curto prazo.
3. **Cautela metodológica:** Como as mortes por intervenção policial compõem o próprio denominador da letalidade violenta (definição oficial do ISP-RJ), a leitura combinada de nível e variação anual demonstra que mudanças na proporção de MDIP refletem sobretudo **alterações na composição da violência letal** — ora acompanhando repiques anuais, ora evidenciando maior resistência da letalidade policial diante da queda dos demais crimes contra a vida.

## Pergunta 7 — Correlação Temporal entre Policiais Mortos em Serviço e Letalidade Violenta Subsequente

> **Pergunta de domínio:** *Existe correlação temporal observável (como picos e vales simultâneos) entre a quantidade de policiais mortos em serviço e a escalada da letalidade violenta geral nos meses subsequentes?*

### 7.1 Transformação e Preparação

Integramos mensalmente `df_policial` (`evolucao_policial_morto_servico_mensal`), somando militares e civis mortos em serviço (`policiais_mortos_servico = quantidade_policial_militar_morto_servico + quantidade_policial_civil_morto_servico`), à série estadual mensal `df_evol_uf` (`quantidade_letalidade_violenta` e `quantidade_homicidio_intervencao_policial`).

**Rigor estatístico na análise de defasagens (lags 0, 1, 2 e 3 meses):**
Em séries temporais longas, duas variáveis que compartilham uma tendência secular comum tendem a apresentar correlação positiva elevada em nível mesmo sem resposta dinâmica imediata de um mês para o outro. Para distinguir **tendência compartilhada** de **escalada no mês subsequente**, calculamos a correlação cruzada $\text{Corr}(X_t, Y_{t+k})$ para $k \in \{0, 1, 2, 3\}$ meses sob dois regimes:
1. **Em nível** (série mensal observada);
2. **Em primeira diferença mensal** ($\Delta X_t = X_t - X_{t-1}$ vs. $\Delta Y_{t+k} = Y_{t+k} - Y_{t+k-1}$), que remove a tendência de longo prazo e isola choques de curto prazo, reportando também a banda aproximada de referência exploratória sob hipótese de ruído branco ($\pm 1{,}96 / \sqrt{n_{\text{diff}}}$, sem pré-branqueamento formal).

```python
exigir_colunas(
    df_policial,
    ["ano", "mes", "quantidade_policial_militar_morto_servico", "quantidade_policial_civil_morto_servico"],
    "df_policial",
)
exigir_colunas(
    df_evol_uf,
    ["ano", "mes", "quantidade_letalidade_violenta", "quantidade_homicidio_intervencao_policial"],
    "df_evol_uf",
)

pol_mensal = (
    df_policial.groupby(["ano", "mes"], as_index=False)[
        ["quantidade_policial_militar_morto_servico", "quantidade_policial_civil_morto_servico"]
    ]
    .sum()
)
pol_mensal["policiais_mortos_servico"] = (
    pol_mensal["quantidade_policial_militar_morto_servico"]
    + pol_mensal["quantidade_policial_civil_morto_servico"]
)

serie_p7 = (
    pol_mensal.merge(
        df_evol_uf[["ano", "mes", "quantidade_letalidade_violenta", "quantidade_homicidio_intervencao_policial"]],
        on=["ano", "mes"],
        how="inner",
        validate="one_to_one",
    )
    .sort_values(["ano", "mes"])
    .reset_index(drop=True)
)
serie_p7["data"] = pd.to_datetime({"year": serie_p7["ano"], "month": serie_p7["mes"], "day": 1})

serie_p7["mm6_policiais"] = serie_p7["policiais_mortos_servico"].rolling(6, center=True).mean()
serie_p7["mm6_letalidade"] = serie_p7["quantidade_letalidade_violenta"].rolling(6, center=True).mean()

serie_p7["d_policiais"] = serie_p7["policiais_mortos_servico"].diff()
serie_p7["d_letalidade"] = serie_p7["quantidade_letalidade_violenta"].diff()
serie_p7["d_mdip"] = serie_p7["quantidade_homicidio_intervencao_policial"].diff()

lags = [0, 1, 2, 3]
resultados_lag = []
for k in lags:
    r_let_nivel = serie_p7["policiais_mortos_servico"].corr(serie_p7["quantidade_letalidade_violenta"].shift(-k))
    r_let_diff = serie_p7["d_policiais"].corr(serie_p7["d_letalidade"].shift(-k))
    r_mdip_nivel = serie_p7["policiais_mortos_servico"].corr(serie_p7["quantidade_homicidio_intervencao_policial"].shift(-k))
    r_mdip_diff = serie_p7["d_policiais"].corr(serie_p7["d_mdip"].shift(-k))
    n_eff_diff = int(
        pd.concat([serie_p7["d_policiais"], serie_p7["d_letalidade"].shift(-k)], axis=1)
        .dropna()
        .shape[0]
    )
    resultados_lag.append({
        "lag_meses": k,
        "n_efetivo_diff": n_eff_diff,
        "r_letalidade_nivel": r_let_nivel,
        "r_letalidade_diff": r_let_diff,
        "r_mdip_nivel": r_mdip_nivel,
        "r_mdip_diff": r_mdip_diff,
    })

resumo_lags_p7 = pd.DataFrame(resultados_lag).round(4)
n_meses_p7 = len(serie_p7)
n_diff_p7 = int(serie_p7["d_policiais"].notna().sum())
limiar_sig_95 = 1.96 / np.sqrt(n_diff_p7)

print(
    f"Série mensal integrada P7: {serie_p7['data'].min().strftime('%Y-%m')} a {serie_p7['data'].max().strftime('%Y-%m')} "
    f"(n = {n_meses_p7} meses; n_diff = {n_diff_p7} | referência exploratória ±1,96/√n_diff: ±{limiar_sig_95:.3f})"
)
print(
    f"Total de policiais mortos em serviço no período: {int(serie_p7['policiais_mortos_servico'].sum())} "
    f"(pico mensal: {int(serie_p7['policiais_mortos_servico'].max())} em "
    f"{serie_p7.loc[serie_p7['policiais_mortos_servico'].idxmax(), 'data'].strftime('%Y-%m')})"
)
display(resumo_lags_p7)
```

### 7.3 Análise e Achados

1. **Co-movimento em nível reflete ciclos históricos compartilhados (2003–2026):** Ao longo dos **284 meses** integrados (janeiro de 2003 a agosto de 2026), registraram-se **596 policiais mortos em serviço** (pico mensal de **10** em outubro de 2004). Na série em nível, a correlação com a letalidade violenta estadual mantém-se moderada e praticamente estável em todas as defasagens — **lag 0: $r = +0{,}441$**; **lag 1: $r = +0{,}415$**; **lag 2: $r = +0{,}389$**; **lag 3: $r = +0{,}447$** (e, contra MDIP em nível, entre $+0{,}107$ e $+0{,}163$). Essa estabilidade entre lags 0 e 3 é a assinatura típica de uma **tendência secular comum** (ambas as séries partiram de patamares elevados nos anos 2000 e declinaram ao longo de duas décadas), e não de uma reação específica no mês imediatamente posterior.
2. **Ausência de escalada sistemática em primeiras diferenças (choques de curto prazo):** Ao remover a tendência secular por meio de primeiras diferenças mensais ($\Delta X_t$ vs. $\Delta Y_{t+k}$), a correlação desaparece no mês contemporâneo e no mês seguinte: para a letalidade violenta total, **lag 0 = $+0{,}059$** e **lag 1 = $+0{,}005$** (muito abaixo da banda de significância aproximada de $\pm 0{,}116$); contra mortes por intervenção policial especificamente, **todos os quatro lags permanecem dentro da faixa de ruído branco** (**lag 0 = $+0{,}026$**, **lag 1 = $+0{,}015$**, **lag 2 = $-0{,}093$**, **lag 3 = $+0{,}023$**). As oscilações modestas e de sinais opostos em lag 2 ($-0{,}162$) e lag 3 ($+0{,}147$) na letalidade total — que excedem a referência exploratória de $\pm 0{,}116$ — refletem reversão à média e autocorrelação/sazonalidade residual após diferenciação (sem pré-branqueamento), não uma escalada sustentada.
3. **Interpretação estatística rigorosa (sem inferência causal):** Os dados estaduais mensais **não corroboram** uma relação causal direta de que meses com mais policiais mortos em serviço provoquem automaticamente uma escalada generalizada da letalidade violenta ou da MDIP nos meses $t+1$ a $t+3$. Eventuais respostas operacionais após mortes de agentes tendem a ocorrer em escalas **microlocais** (CISP/AISP) e janelas de **dias ou semanas**, diluindo-se na agregação mensal estadual.

# Conclusão

## Síntese Integrada dos Achados

A análise longitudinal e georreferenciada das bases oficiais do ISP-RJ demonstra que a segurança pública fluminense não pode ser compreendida por um único indicador agregado. Embora a taxa estadual de letalidade violenta tenha recuado expressivamente no longo prazo (de 60,3 em 1991 e 54,3 em 2003 para 25,5 por 100 mil hab. em 2022) e o volume total de armas apreendidas tenha caído 44,4% entre 2007 e 2024, **a violência letal transformou-se qualitativamente e redistribuiu-se no território**:

- **Maior poder de fogo em menos armas (P1):** os fuzis mais que triplicaram em volume (+242,1%, de 214 para 732) e sextuplicaram sua participação no arsenal apreendido (1,93% → 11,90%), acompanhados pela predominância das pistolas (50,96% em 2024) sobre os revólveres (25,51%).
- **Convergência regional e interiorização relativa (P4):** a queda acentuada na Capital (−66,7% entre 2003 e 2022) e em Grande Niterói (−52,7%) contrastou com a maior rigidez no Interior (−24,6%), invertendo o ordenamento histórico frente à Capital em 2022 (Interior 26,0 vs. Capital 19,4/100 mil), enquanto a Baixada Fluminense manteve o maior patamar proporcional do estado (35,9/100 mil).
- **Fenômenos espacialmente distintos (P2 e P5):** a letalidade violenta e a letalidade policial concentram-se em um corredor metropolitano contínuo fortemente associado à apreensão de fuzis (15 das 39 AISPs no Cluster 1; Spearman $\rho = 0{,}834$), ao passo que o **feminicídio** exibe geografia própria (6 das 10 maiores taxas no Interior; $r_{\text{espacial}} = 0{,}354$) e trajetória temporal ascendente em 2017–2025 ($r = -0{,}869$ frente à letalidade geral).
- **Sazonalidade, composição municipal e dinâmica de curto prazo (P3, P6 e P7):** todas as regiões compartilham picos de intensidade diária em fevereiro–março e vales em julho–agosto (amplitude estadual de 22,2 pontos); nos grandes municípios metropolitanos (2014–2024), a participação de MDIP chegou a 55,7% em Niterói e 48,7% em São Gonçalo mesmo em contexto de queda da letalidade geral; por fim, a associação entre policiais mortos em serviço e letalidade estadual ao longo de 284 meses decorre de tendência histórica compartilhada, sem evidência de escalada sistemática no mês seguinte em primeiras diferenças ($r_{\text{lag 1}} = +0{,}005$).

## Recomendações

Com base nas evidências empíricas levantadas ao longo das sete perguntas, sugerimos quatro frentes prioritárias de atuação e aprimoramento institucional:

1. **Focalização territorial do controle de armamento pesado (P1 e P5):** priorizar inteligência policial, rastreamento de rotas de entrada e operações integradas contra armas de guerra nas **15 AISPs do Cluster 1** (corredor metropolitano e Angra dos Reis), onde altas taxas de MDIP e apreensões de fuzis se retroalimentam.
2. **Políticas autônomas de prevenção ao feminicídio no Interior (P2):** desacoplar o enfrentamento ao feminicídio da agenda de confronto armado urbano, ampliando delegacias especializadas (DEAMs), patrulhas Maria da Penha e redes de acolhimento nos polos do **Interior** que lideram as taxas proporcionais (como Santo Antônio de Pádua, Campos dos Goytacazes e Itaperuna).
3. **Planejamento operacional sensível à sazonalidade (P3 e P4):** reforçar preventivamente o efetivo e o monitoramento no **primeiro quadrimestre (janeiro a abril)** — em especial fevereiro e março — e ampliar investimentos de policiamento no **Interior**, cuja queda histórica de letalidade foi substancialmente mais lenta (−24,6%) que a da Capital (−66,7%).
4. **Aprimoramento da infraestrutura de dados abertos (P1, P2 e P6):** publicar séries populacionais atualizadas pós-2022 por CISP/AISP (alinhadas ao Censo IBGE 2022) e desagregadas por sexo, além de explicitar vigências históricas na tabela de relacionamento CISP–AISP–RISP, eliminando a necessidade de propagações (`ffill`) ou aproximações indiretas.

## Limitações Metodológicas

1. **Cobertura temporal das séries populacionais:** a base `PopulacaoEvolucaoMensalCisp` e a população estadual de `df_letalidade` encerram-se em **2022**. Assim, análises de taxas por AISP limitam-se a `2003–2022` (e `2017–2022` no cruzamento com feminicídio), as taxas estaduais de armas em 2023–2024 dependem de propagação (`ffill`) da população de 2022 e a seleção municipal em P6 recorre à inversão da taxa oficial de registros de ocorrência.
2. **Aproximação da população feminina e amostras reduzidas:** diante da ausência de séries populacionais por sexo por CISP/AISP, a taxa de feminicídio assume 50% de população feminina em cada AISP; além disso, territórios menores apresentam contagens absolutas baixas, o que aumenta a sensibilidade das taxas proporcionais.
3. **Ausência de vigência temporal na tabela CISP–AISP–RISP e natureza administrativa dos registros:** o arquivo `relacao_cisp_aisp_risp` reflete a divisão territorial mais recente, razão pela qual as taxas históricas utilizam a composição mensal observada por CISP. Todas as métricas derivam de Registros de Ocorrência (ROs), sujeitos a mudanças de tipificação legal (como a Lei nº 13.104/2015), variações de subnotificação e intensidade operacional — impondo interpretação descritiva e exploratória, nunca causalidade isolada.
